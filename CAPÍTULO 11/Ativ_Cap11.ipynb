#Enunciado 10:
Peça a quantidade de alunos da turma e, para cada aluno, o nome e quatro notas (P1, P2,
P3 e Projeto, com pesos 30%, 30%, 20% e 20%). Calcule a média ponderada de cada
aluno, classifique-o (Aprovado se média >= 7, Recuperação se média >= 5, Reprovado nos
demais casos), conte quantos alunos ficaram em cada situação, e ao final exiba a média
geral da turma, o aluno com a maior média e o aluno com a menor média.

```python
quant_aluno = int(input("Informe a quantidade de alunos da turma: "))
aprovados = 0
recuperacao = 0
reprovados = 0
maior_media = 0
menor_media = 10
total_media = 0
media_turma = total_media / quant_aluno
nome_maior_media = ""
nome_menor_media = ""


for i in range(quant_aluno):
  nome = input(f"Informe o nome do aluno{i}: ")
  p1 = float(input("Informe a nota P1: "))
  p2 = float(input("Informe a nota P2: "))
  p3 = float(input("Informe a nota P3: "))
  projeto = float(input("Informe a nota do Projeto: "))
  media = (p1 * 0.3) + (p2 * 0.3) + (p3 * 0.2) + (projeto * 0.2)
  total_media += media
  if media >= 7:
    situacao = "Aprovado"
    aprovados += 1
    print(f"O aluno {nome} foi {situacao} com a média {media}")
    if media > maior_media:
      maior_media = media
      nome_maior_media = nome
    elif media < menor_media:
      menor_media = media
      nome_menor_media = nome
  elif media >= 5:
    situacao = "Recuperação"
    recuperacao += 1
    print(f"O aluno {nome} ficou de  {situacao} com a média {media}")
    if media > maior_media:
      maior_media = media
      nome_maior_media = nome
    elif media < menor_media:
      menor_media = media
      nome_menor_media = nome
  else:
    situacao = "Reprovado"
    reprovados += 1
    print(f"O aluno {nome} foi {situacao} com a média {media}")
    if media > maior_media:
      maior_media = media
      nome_maior_media = nome
    elif media < menor_media:
      menor_media = media
      nome_menor_media = nome
print(f"A média geral da turma é {media_turma}")
print(f"O aluno com a maior média é {nome_maior_media} com a média {maior_media}")
print(f"O aluno com a menor média é {nome_menor_media} com a média {menor_media}")
```

#Enunciado 11:
Para 10 vendedores, peça o valor vendido por cada um. Ao final, exiba o total vendido, a
média de vendas, quantos vendedores bateram a meta de R$ 5.000,00 e o percentual
correspondente.

```python
total_vendido = 0
meta = 5000
vendedores_meta = 0
for i in range(1,11):
  valor_vendido = float(input(f"Informe o valor vendido pelo vendedor {i}: "))
  total_vendido += valor_vendido
  if valor_vendido >= meta:
    vendedores_meta += 1
  else:
    total_vendido += valor_vendido
media_vendida = total_vendido / 10
percentual_meta = (vendedores_meta / 10) * 100
print(f"O total vendido foi de {total_vendido}")
print(f"A média de vendas foi de {media_vendida}")
print(vendedores_meta)
print(percentual_meta)
```

#Enunciado 12:
Peça um número inteiro n e calcule n! (fatorial de n), que é o produto de todos os inteiros de 1 até n.

```python
fatorial = 1
numero = int(input("Informe um número inteiro: "))
num = numero + 1
for i in range(1, num):
  fatorial = fatorial * i
print(f"O fatorial de {numero} é {fatorial}")
```

# Enunciado 14:
Peça a uma IA para gerar o programa do Exercício 10 (Processamento de turma com
nomes). Avalie a resposta usando o checklist abaixo, e escreva uma análise da auditoria.
image.png

```python
#10º Resposta:
n_alunos = int(input('Quantidade de alunos: '))

if n_alunos <= 0:
    print('A turma precisa ter pelo menos 1 aluno.')
else:
    soma_medias = 0
    aprovados = 0
    recuperacao = 0
    reprovados = 0
    maior_media = -1          # qualquer média real (0 a 10) será maior
    menor_media = 11          # qualquer média real será menor
    nome_maior = ''
    nome_menor = ''

    for i in range(1, n_alunos + 1):
        print(f'\n--- Aluno {i} ---')
        nome = input('Nome: ')
        p1 = float(input('P1: '))
        p2 = float(input('P2: '))
        p3 = float(input('P3: '))
        projeto = float(input('Projeto: '))

        media = p1 * 0.30 + p2 * 0.30 + p3 * 0.20 + projeto * 0.20
        soma_medias += media

        if media >= 7:
            situacao = 'Aprovado'
            aprovados += 1
        elif media >= 5:
            situacao = 'Recuperação'
            recuperacao += 1
        else:
            situacao = 'Reprovado'
            reprovados += 1

        if media > maior_media:
            maior_media = media
            nome_maior = nome
        if media < menor_media:
            menor_media = media
            nome_menor = nome

        print(f'{nome}: média {media:.2f} — {situacao}')

    media_geral = soma_medias / n_alunos
    print('\n===== RESUMO DA TURMA =====')
    print(f'Aprovados: {aprovados}')
    print(f'Recuperação: {recuperacao}')
    print(f'Reprovados: {reprovados}')
    print(f'Média geral da turma: {media_geral:.2f}')
    print(f'Maior média: {nome_maior} ({maior_media:.2f})')
    print(f'Menor média: {nome_menor} ({menor_media:.2f})')
```

#Enunciado 15:
O código abaixo soma todos os números pares entre 1 e 100 e conta quantos foram
encontrados. Explique linha a linha o que ele faz.
soma = 0
contador = 0
for i in range(1, 101):
if i % 2 == 0:
soma += i
contador += 1
print(soma)
print(contador)

##15º Resposta:
1. `soma = 0` — cria o acumulador da soma, começando no valor neutro 0.
2. `contador = 0` — cria o contador de pares, também começando em 0.
3. `for i in range(1, 101):` — repete o bloco para cada `i` de 1 até 100 (o 101 não entra).
4. `if i % 2 == 0:` — testa se o resto da divisão de `i` por 2 é zero, ou seja, se `i` é par.
5. `soma += i` — se for par, soma `i` ao acumulador.
6. `contador += 1` — e aumenta em 1 a quantidade de pares encontrados.
7. `print(soma)` — depois do loop, exibe a soma: **2550**.
8. `print(contador)` — exibe a quantidade de pares: **50**.

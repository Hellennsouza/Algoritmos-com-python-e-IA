```python
nota = 0
soma = 0
qtd = 0
while nota != -1:
  nota = float(input("Digite a nota: "))
  while nota < -1 or nota > 10:
    nota = float(input("Digite a nota: "))

  if nota != -1:
    soma = soma + nota
    qtd = qtd + 1
media = soma/qtd
print(f"Média: {media:6.2f}")
```

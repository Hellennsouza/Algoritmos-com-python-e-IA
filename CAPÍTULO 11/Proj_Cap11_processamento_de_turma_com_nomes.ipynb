#**Projeto do capítulo — Processamento de turma com nomes (versão expandida)**
Peça a quantidade de alunos e, para cada um, leia o nome e quatro notas: P1 (30%), P2 (30%), P3 (20%) e Projeto (20%). Calcule a média ponderada, classifique o aluno (Aprovado ≥ 7, Recuperação ≥ 5, Reprovado < 5), conte cada situação e, ao final, exiba um resumo com total de alunos, média geral, percentuais e o melhor e o pior desempenho.

```python
#Projeto — Resposta:
n_alunos = int(input('Quantos alunos há na turma? '))

if n_alunos <= 0:
    print('Erro: informe uma quantidade de alunos maior que zero.')
else:
    # acumuladores e contadores inicializados ANTES do loop
    soma_medias = 0.0
    aprovados = 0
    recuperacao = 0
    reprovados = 0
    maior_media = -1
    menor_media = 11
    nome_maior = ''
    nome_menor = ''

    print('\n=== Processamento da Turma ===\n')

    for i in range(n_alunos):
        print(f'--- Aluno {i + 1} de {n_alunos} ---')
        nome = input('  Nome: ')
        p1 = float(input('  Nota P1 (30%): '))
        p2 = float(input('  Nota P2 (30%): '))
        p3 = float(input('  Nota P3 (20%): '))
        projeto = float(input('  Nota Projeto (20%): '))


        media = (p1 * 0.30) + (p2 * 0.30) + (p3 * 0.20) + (projeto * 0.20)
        soma_medias += media

        if media >= 7:
            situacao = 'Aprovado'
            aprovados += 1
        elif media >= 5:
            situacao = 'Recuperação'
            recuperacao += 1
        else:
            situacao = 'Reprovado'
            reprovados += 1

        # dois if separados: maior e menor são atualizados de forma independente
        if media > maior_media:
            maior_media = media
            nome_maior = nome
        elif media < menor_media:
            menor_media = media
            nome_menor = nome

        print(f'  Média: {media:.2f} | Situação: {situacao}\n')

    media_turma = soma_medias / n_alunos

    print('=' * 40)
    print(' RESUMO DA TURMA')
    print('=' * 40)
    print(f'Total de alunos: {n_alunos}')
    print(f'Média geral: {media_turma:.2f}')
    print(f'Aprovados: {aprovados} ({aprovados / n_alunos:.0%})')
    print(f'Em recuperação: {recuperacao} ({recuperacao / n_alunos:.0%})')
    print(f'Reprovados: {reprovados} ({reprovados / n_alunos:.0%})')
    print(f'Melhor desempenho: {nome_maior} ({maior_media:.2f})')
    print(f'Menor desempenho: {nome_menor} ({menor_media:.2f})')
    print('=' * 40)
```

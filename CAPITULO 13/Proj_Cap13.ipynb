#**Projeto do capítulo — Mini ETL textual de cadastro de clientes**
Cada linha do cadastro tem o formato `nome ; cpf ; email ; cidade`, com dados inconsistentes (maiúsculas, espaços extras, CPFs com e sem pontuação, e-mails inválidos). Para cada linha: separar os campos, limpar e padronizar, validar CPF e e-mail, decidir a situação (Válido ou Revisar) e exibir um relatório.

```python
#Projeto — Resposta:
# Extract — em um sistema real, as linhas viriam de um arquivo
linhas = [
    'ANA SILVA ; 123.456.789-00 ; ANA@EMAIL.COM ; sorocaba',
    'bruno costa ; 98765432100 ; bruno@email ; VOTORANTIM',
    ' Carla Lima ; 111.222.333-44 ; carla@email.com ; são paulo',
    'Dados Incompletos ; 123 ; sem_email',
    'DIEGO  souza ; 555.666.777-8 ; diego@@email.com ; itu',
]

validos = 0
revisar = 0

print(f"{'Nome':<18} {'CPF':<16} {'E-mail':<18} {'Cidade':<12} Situação")
print('-' * 76)

for linha in linhas:
    campos = linha.split(';')

    if len(campos) != 4:
        print(f"{'Linha incompleta':<18} {'-':<16} {'-':<18} {'-':<12} Revisar")
        revisar += 1
    else:
        # Transform — limpeza e padronização
        nome = ' '.join(campos[0].split()).title()   # também remove espaços duplos no meio
        cpf_bruto = campos[1].strip()
        email = campos[2].strip().lower()
        cidade = campos[3].strip().title()

        # CPF: remove pontuação e valida o formato básico
        cpf_limpo = cpf_bruto.replace('.', '').replace('-', '')
        cpf_ok = cpf_limpo.isdigit() and len(cpf_limpo) == 11

        # E-mail: um @, ponto depois do @, sem espaços, mínimo de 5 caracteres
        um_arroba = email.count('@') == 1
        pos_arroba = email.find('@')
        ponto_ok = pos_arroba != -1 and '.' in email[pos_arroba + 1:]
        email_ok = um_arroba and ponto_ok and ' ' not in email and len(email) >= 5

        if cpf_ok and email_ok:
            situacao = 'Válido'
            validos += 1
        else:
            situacao = 'Revisar'
            revisar += 1

        # Load (exibição) — CPF formatado só quando é válido
        if cpf_ok:
            cpf_exib = f'{cpf_limpo[:3]}.{cpf_limpo[3:6]}.{cpf_limpo[6:9]}-{cpf_limpo[9:]}'
        else:
            cpf_exib = cpf_limpo

        print(f'{nome:<18} {cpf_exib:<16} {email:<18} {cidade:<12} {situacao}')

print('-' * 76)
print(f'Registros válidos: {validos}')
print(f'Registros para revisão: {revisar}')
```

##Projeto — Por que cada registro ficou assim:
| Registro | Situação | Motivo |
|---|---|---|
| Ana Silva | Válido | CPF com 11 dígitos e e-mail completo (padronizado para minúsculas) |
| Bruno Costa | Revisar | CPF ok, mas `bruno@email` não tem ponto depois do @ |
| Carla Lima | Válido | Espaço inicial removido com `strip()` |
| Linha incompleta | Revisar | Só 3 campos depois do `split(';')` |
| Diego Souza | Revisar | CPF com 10 dígitos e e-mail com dois @ |

**Diferença em relação à solução do professor:** não usei `continue` (ainda não estudado); a linha incompleta é tratada com `if/else`. Também acrescentei o caso do Diego para testar CPF curto, @ duplicado e espaços duplos no meio do nome (`' '.join(texto.split())`).

```python
frase = input("Informe a frase:")
quant_vogal = 0
consoantes = 0
for letra in frase:
    if letra.upper() in 'AEIOUÁÉÍÓÚÀÈÌÒÙÂÔÊÎÛÃÕ':
        quant_vogal += 1
    else:
        consoantes += 1
print(f'A frase possui', quant_vogal, 'vogais e',consoantes, 'consoantes')
```

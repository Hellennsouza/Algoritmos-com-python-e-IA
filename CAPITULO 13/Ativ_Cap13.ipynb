#**8º Enunciado:**
Identifique e corrija a comparação de string sem normalização prévia:

```python
estado = input('Estado: ')
if estado == 'SP':
    print('São Paulo')
else:
    print('Outro estado')
```

##8º Resposta:
A comparação `estado == 'SP'` só é verdadeira se o texto for literalmente `'SP'`. Se o usuário digitar `'sp'`, `' SP'` ou `'SP '`, o resultado é `False`, e o espaço é invisível na tela. A correção é normalizar a entrada antes de comparar: `strip()` tira os espaços das pontas e `upper()` coloca tudo em maiúsculas.

````python
#8º Resposta (corrigido):
estado = input('Estado: ').strip().upper()
if estado == 'SP':
    print('São Paulo')
else:
    print('Outro estado')
````

#**9º Enunciado:**
Explique por que o código abaixo exibe 'ANA' e não 'ana', e corrija.

```python
nome = "ANA"
nome.lower()
print(nome)
```

##9º Resposta:
Strings em Python são **imutáveis**: `nome.lower()` cria uma *nova* string `'ana'`, mas o resultado não é guardado em lugar nenhum e se perde. A variável `nome` continua com `'ANA'`. É preciso atribuir o retorno do método:

````python
#9º Resposta (corrigido):
nome = "ANA"
nome = nome.lower()        # reatribui o resultado
print(nome)

nome_original = "ANA"      # ou guarda em outra variável e preserva o original
nome_minusculo = nome_original.lower()
print(nome_original, nome_minusculo)
````

#**10º Enunciado:**
Corrija o código para que não trave quando o usuário pressionar Enter sem digitar nada.

```python
texto = input('Texto: ')
palavras = texto.split()
maior = max(palavras, key=len)
print(f'Palavra mais longa: {maior}')
```

##10º Resposta:
Com texto vazio, `split()` devolve uma lista vazia `[]` e `max()` gera `ValueError: max() arg is an empty sequence`. Basta verificar se há palavras antes de chamar `max()`:

````python
#10º Resposta (corrigido):
texto = input('Texto: ')
palavras = texto.split()

if len(palavras) == 0:
    print('Nenhuma palavra foi digitada.')
else:
    maior = max(palavras, key=len)
    print(f'Palavra mais longa: {maior}')
````

#**11º Enunciado:**
Leia um parágrafo e exiba seis estatísticas: número de caracteres com espaços, número de caracteres sem espaços, número de palavras, número de frases (contadas por pontuação final), número de vogais e a palavra mais longa.

````python
#11º Resposta:
texto = input('Digite um parágrafo: ')

caracteres_com = len(texto)
caracteres_sem = len(texto.replace(' ', ''))

palavras = texto.split()
num_palavras = len(palavras)

num_frases = texto.count('.') + texto.count('!') + texto.count('?')

vogais = 'aeiouáéíóúàâêôãõ'
num_vogais = 0
for caractere in texto.lower():
    if caractere in vogais:
        num_vogais += 1

if num_palavras > 0:
    mais_longa = max(palavras, key=len)
else:
    mais_longa = '(nenhuma)'

print('=== Análise do Texto ===')
print(f'Caracteres (com espaços): {caracteres_com}')
print(f'Caracteres (sem espaços): {caracteres_sem}')
print(f'Palavras: {num_palavras}')
print(f'Frases: {num_frases}')
print(f'Vogais: {num_vogais}')
print(f'Palavra mais longa: {mais_longa}')
````

#**12º Enunciado:**
Leia uma linha no formato produto,categoria,preco,quantidade, processe cada campo e exiba o nome do produto, a categoria e o valor total (preço × quantidade).

````python
#12º Resposta:
linha = input('produto,categoria,preco,quantidade: ')
campos = linha.split(',')

if len(campos) != 4:
    print('Linha inválida. Formato esperado: produto,categoria,preco,quantidade')
else:
    produto = campos[0].strip().title()
    categoria = campos[1].strip().title()
    preco = float(campos[2].strip())
    quantidade = int(campos[3].strip())
    total = preco * quantidade

    print(f'Produto:   {produto}')
    print(f'Categoria: {categoria}')
    print(f'Total:     R$ {total:.2f}')
````

#**13º Enunciado:**
Peça à IA um programa para validar CPF no formato 000.000.000-00, identifique os métodos de string usados, pesquise-os em docs.python.org e avalie se a solução está correta, legível e adequada para iniciante.

````python
#13º Resposta — solução gerada pela IA (Claude) para auditoria:
import re

cpf = input('Digite o CPF (000.000.000-00): ')
if re.fullmatch(r'\d{3}\.\d{3}\.\d{3}-\d{2}', cpf):
    numeros = cpf.replace('.', '').replace('-', '')
    if numeros == numeros[0] * 11:
        print('CPF inválido (todos os dígitos iguais).')
    else:
        print('Formato de CPF válido.')
else:
    print('Formato inválido. Use 000.000.000-00.')
````

|---|---|

| Aceita CPF sem formatação? | **Não**: `12345678900` é recusado, porque a regex exige pontos e traço |
| Trata entrada vazia? | Sim, sem travar: a regex não casa e aparece "Formato inválido" |
| Valida dígitos verificadores? | Não; só recusa CPFs com todos os dígitos iguais (bom detalhe extra) |
| Consigo explicar linha a linha? | Não a regex `\d{3}\.\d{3}…` sem consultar a documentação |

**Métodos pesquisados em docs.python.org:** `str.replace(old, new)` devolve uma cópia com as trocas; `str.isdigit()` retorna `True` se todos os caracteres forem dígitos e houver pelo menos um; `re.fullmatch(padrão, texto)` só casa se o texto inteiro seguir o padrão.

**Conclusão:** a solução funciona para o formato pedido, mas usa regex e recusa CPF sem pontuação. Para o nível do capítulo, o código de referência (`replace()` + `isdigit()` + `len() == 11`) é mais legível e mais flexível. Aproveitaria da IA a ideia de recusar CPFs com todos os dígitos iguais.

#**2º Enunciado** — Tipos no Colab

Crie uma variável de cada um dos quatro tipos primitivos (str, int, float, bool) e use

#**3º Enunciado:** — Idade no próximo ano

Peça a idade do usuário e exiba uma mensagem informando a idade que ele terá no próximo ano.
Como pensar antes de programar
input() sempre devolve texto — mesmo quando o usuário digita números. Para fazer uma
conta com a idade, é preciso transformar esse texto em número inteiro com int() antes de
somar.

```python
#3ºRespostas
idade = int(input("Qual a sua idade?"))
print(f'Em um ano você terá {idade + 1} anos')
```

#**5º Enunciado:** — Ticket médio de vendas

Peça o faturamento total e a quantidade de pedidos de uma loja, e calcule o ticket médio
(faturamento dividido pela quantidade de pedidos), exibindo o resultado com duas casas
decimais.

```python
fat_total = float(input("Qual o faturamento total?"))
num_pedidos = int(input("Quantos pedidos foram feitos?"))
ticket_medio = fat_total / num_pedidos
print(f'O ticket médio é de R${ticket_medio:.2f}')
```

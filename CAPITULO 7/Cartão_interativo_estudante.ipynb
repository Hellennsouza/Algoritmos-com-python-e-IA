```python
# PROJETO: Cartão interativo do estudante
# Capítulo 7 — Variáveis, tipos, input() e print()
# COLETA
# RA fica como str: não será usado em cálculo, pode ter zeros
nome = input('Nome completo: ')
ra = input('RA: ')
cidade = input('Cidade: ')
idade = int(input('Idade: '))
semestre = int(input('Semestre atual: '))
media_atual = float(input('Média atual: '))
# PROCESSAMENTO
# Estimativa baseada no ano atual (2026)
ano_nascimento = 2026 - idade
# Semestre seguinte ao informado
proximo_semestre = semestre + 1
# EXIBIÇÃO
print('==============================')
print(' CARTÃO DO ESTUDANTE')
print('==============================')
print(f'Nome: {nome}')
print(f'RA: {ra}')
print(f'Cidade: {cidade}')
print(f'Idade: {idade} anos')
print(f'Nasc. aprox.: {ano_nascimento}')
print(f'Semestre atual: {semestre}')
print(f'Próximo semestre: {proximo_semestre}')
print(f'Média atual: {media_atual:.2f}')
print('==============================')
```

#**1º Enunciado:**
Escreva um programa que imprima os números de 1 a 10 usando while.

````python
#1º Resposta:
i = 1
while i <= 10:
    print(i)
    i += 1          # sem esta atualização o loop seria infinito
````

#**2º Enunciado:**
Escreva um programa que imprima os números de 10 a 1 usando while e, ao final, mostre a palavra 'Fim'.

````python
#2º Resposta:
i = 10
while i >= 1:
    print(i)
    i -= 1
print('Fim')
````

#**6º Enunciado:**
Crie um menu com as opções 1 (área do círculo), 2 (área do retângulo) e 0 (sair). O menu deve repetir até o usuário escolher sair.

````python
#6º Resposta:
opcao = -1
while opcao != 0:
    print('\n===== MENU =====')
    print('1 - Área do círculo')
    print('2 - Área do retângulo')
    print('0 - Sair')
    opcao = int(input('Escolha: '))

    if opcao == 1:
        raio = float(input('Raio: '))
        area = 3.14159 * raio ** 2
        print(f'Área do círculo: {area:.2f}')
    elif opcao == 2:
        base = float(input('Base: '))
        altura = float(input('Altura: '))
        print(f'Área do retângulo: {base * altura:.2f}')
    elif opcao == 0:
        print('Encerrando o programa.')
    else:
        print('Opção inválida.')
````

#**8º Enunciado:**
O código abaixo nunca termina. Identifique o problema, explique por que ele causa loop infinito e mostre a correção.

```python
i = 1
while i <= 5:
    print(i)
    # i += 1 foi esquecido!
```

##8º Resposta:
A condição `i <= 5` depende de `i`, mas nada dentro do loop altera `i`. Ele vale 1 para sempre, a condição é sempre `True` e o programa imprime 1 infinitamente. A correção é atualizar a variável da condição dentro do bloco:

````python
#8º Resposta (corrigido):
i = 1
while i <= 5:
    print(i)
    i += 1
````

#**9º Enunciado:**
Explique por que a validação abaixo nunca funciona e mostre a correção.

```python
nota = float(input('Nota: '))
while nota < 0 and nota > 10:
    nota = float(input('Nota invalida. Digite novamente: '))
```

##9º Resposta:
Com `and`, as duas condições precisariam ser verdadeiras ao mesmo tempo: a nota seria menor que 0 **e** maior que 10, o que é impossível. A condição é sempre `False`, o `while` nunca roda e qualquer valor é aceito. O certo é `or`: basta a nota estar fora da faixa por um dos lados para pedir de novo.

````python
#9º Resposta (corrigido):
nota = float(input('Nota: '))
while nota < 0 or nota > 10:
    nota = float(input('Nota inválida. Digite um valor entre 0 e 10: '))
print(f'Nota aceita: {nota}')
````

#**10º Enunciado:**
Leia valores positivos até o usuário digitar -1. Exiba a soma, a quantidade e a média. O valor -1 não entra no cálculo.

````python
#10º Resposta:
soma = 0
quantidade = 0

valor = float(input('Valor (-1 para encerrar): '))
while valor != -1:
    soma += valor
    quantidade += 1
    valor = float(input('Valor (-1 para encerrar): '))

if quantidade > 0:
    print(f'Soma: {soma:.2f}')
    print(f'Quantidade: {quantidade}')
    print(f'Média: {soma / quantidade:.2f}')
else:
    print('Nenhum valor foi digitado.')
````

#**11º Enunciado:**
O computador sorteia um número entre 1 e 50. O usuário tenta adivinhar sem limite de tentativas. O programa informa se o número secreto é maior ou menor.

````python
#11º Resposta:
import random

numero_secreto = random.randint(1, 50)
tentativas = 0
palpite = 0

while palpite != numero_secreto:
    palpite = int(input('Seu palpite (1 a 50): '))
    tentativas += 1
    if palpite < numero_secreto:
        print('O número secreto é MAIOR.')
    elif palpite > numero_secreto:
        print('O número secreto é MENOR.')

print(f'Acertou! O número era {numero_secreto}. Tentativas: {tentativas}')
````

#**12º Enunciado:**
Adicione pontuação ao jogo. O jogador começa com 100 pontos e perde 10 pontos a cada erro. Há limite de 7 tentativas. Exiba a pontuação ao final.

````python
#12º Resposta:
import random

numero_secreto = random.randint(1, 50)
pontos = 100
tentativas = 0
acertou = False

while tentativas < 7 and not acertou:
    palpite = int(input(f'Tentativa {tentativas + 1} de 7: '))
    tentativas += 1
    if palpite == numero_secreto:
        acertou = True
    else:
        pontos -= 10
        if palpite < numero_secreto:
            print('O número secreto é MAIOR.')
        else:
            print('O número secreto é MENOR.')

if acertou:
    print(f'Acertou em {tentativas} tentativa(s)!')
else:
    print(f'Suas tentativas acabaram. O número era {numero_secreto}.')
print(f'Pontuação final: {pontos}')
````

#**Projeto do capítulo — Jogo de adivinhação com tentativas e opção de jogar novamente**
O computador sorteia um número entre 1 e 100. O jogador tem até 7 tentativas e recebe a dica "maior" ou "menor" a cada chute. Ao final da rodada, o programa informa quantas tentativas foram usadas (ou revela o número, se o jogador perder) e pergunta se ele quer jogar novamente.

```python
#Projeto — Resposta:
import random

LIMITE_TENTATIVAS = 7          # constante: para mudar o limite, mude só aqui
jogar_novamente = 'S'
rodadas = 0
vitorias = 0

# loop externo: controla se o usuário quer jogar de novo
while jogar_novamente.strip().upper() == 'S':
    # variáveis reinicializadas a cada rodada
    numero_secreto = random.randint(1, 100)
    tentativas = 0
    acertou = False
    rodadas += 1

    print('\n' + '=' * 42)
    print(' Adivinhe o número entre 1 e 100!')
    print(f' Você tem {LIMITE_TENTATIVAS} tentativas.')
    print('=' * 42)

    # loop interno: cada tentativa
    while tentativas < LIMITE_TENTATIVAS and not acertou:
        restantes = LIMITE_TENTATIVAS - tentativas
        chute = int(input(f'[{restantes} restante(s)] Seu chute: '))

        # validação: chute fora da faixa não gasta tentativa
        while chute < 1 or chute > 100:
            chute = int(input('  Digite um número entre 1 e 100: '))

        tentativas += 1
        if chute == numero_secreto:
            acertou = True
        elif chute < numero_secreto:
            print(f'  O número é MAIOR que {chute}.')
        else:
            print(f'  O número é MENOR que {chute}.')

    if acertou:
        vitorias += 1
        print(f'\nParabéns! Você acertou em {tentativas} tentativa(s)!')
    else:
        print(f'\nVocê perdeu. O número era {numero_secreto}.')

    jogar_novamente = input('\nJogar novamente? (S/N): ')

print(f'\nRodadas jogadas: {rodadas} | Vitórias: {vitorias}')
print('Obrigado por jogar!')
```

Retorno do Gemini:
1. Os Conceitos Fundamentais
Antes de vermos a transformação no código, precisamos alinhar quatro conceitos essenciais que conectam o Hardware e o Software:

Decimal: É o sistema numérico que usamos no dia a dia (base 10, com dígitos de 0 a 9). Exemplo: o número 65.

Binário: É a linguagem do hardware/processador (base 2, composta apenas por 0 e 1), representando a presença ou ausência de sinal elétrico. Exemplo: 01000001.

Bit: É a menor unidade de informação dentro do computador. Corresponde a um único dígito binário (0 ou 1).

Byte: É um agrupamento de 8 bits. É no espaço de 1 byte que guardamos exatamente 1 caractere da tabela ASCII.

2. Como a Letra Virar Número (Tabela ASCII)
Como o hardware só entende sinais elétricos (binário), foi criada a Tabela ASCII: um "dicionário" que associa cada caractere a um número decimal fixo. O fluxo de transformação ocorre assim:

Você digita a Letra.

O software consulta a Tabela ASCII e encontra o número Decimal correspondente.

O computador converte esse número decimal em Binário (8 bits / 1 byte) para salvar na memória.

Nossos 2 Exemplos:
Exemplo 1 (Letra 'A'):

Letra: 'A'

Código ASCII (Decimal): 65

Em Binário (1 Byte / 8 bits): 01000001

Exemplo 2 (Letra 'B'):

Letra: 'B'

Código ASCII (Decimal): 66

Em Binário (1 Byte / 8 bits): 01000010

3. Na Prática com Python
No Python, usamos duas funções simples para fazer a ponte com a tabela ASCII:

ord(): Recebe 1 caractere e devolve o seu número decimal na ASCII.

chr(): Recebe o número decimal e devolve o caractere correspondente.

Python
# Exemplo 1: Letra 'A'
numero_A = ord('A')  # Retorna 65
letra_A = chr(65)    # Retorna 'A'

# Exemplo 2: Letra 'B'
numero_B = ord('B')  # Retorna 66
letra_B = chr(66)    # Retorna 'B'
4. Erro Comum de Iniciante ⚠️
Um erro muito comum ao usar a função ord() é tentar passar uma palavra inteira (mais de um caractere) de uma só vez:

Python
# ERRO! Isso vai gerar uma falha (TypeError)
print(ord('AB'))
Por que dá erro? A função ord() foi projetada para consultar a tabela ASCII caractere por caractere. Se você passar 'AB', o Python não saberá qual dos dois símbolos converter. Para converter textos maiores, é preciso analisar caractere por caractere (por exemplo, usando um laço de repetição for).

Definição de Algoritmo:

Sequencia finita de passo a passo para resolver um problema.

Dois exemplos de algoritmos do cotidiano:

#Algoritmo para atividade do dia a dia: Lavar a louça

1. Separe a louça suja
2. Jogue o excesso de comida da louça no lixo
3. Coloque a louça suja na pia
4. Pegue um bombril e coloque sobre a buxa de lavar prato
5. Molhe a buxa com o brombril
6. Coloque detergente sobre a buxa de lavar prato
7. Passe a buxa com o bombril em toda area da louça
8. Enxague a louça
9. Verifique que a louça está realmente limpa, caso contrário refaça todos os passos

#Algoritmo para atividade do dia a dia: Limpar a casa

1. Separe a vassoura
2. Varra a casa
3. retire os ciscos com o auxilio de uma pá
4. Encha um balde com agua
5. Coloque os produtos de limpeza no balde
6. Lave o pano de limpar a casa na agua com produto de limpar casa no balde
7. Coloque o pano em um rodo.
8. Passe o pano com o auxilio de um rodo nos comodos internos.
9. Passe o pano com o auxilio de um rodo nos comodos externos.

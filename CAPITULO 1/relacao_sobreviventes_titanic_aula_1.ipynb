```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Carregar a base de dados do Titanic
titanic_df = sns.load_dataset('titanic')

# Mostrar as primeiras linhas do DataFrame para verificar
# print(titanic_df.head())

# Criar um gráfico de contagem para visualizar a relação entre sobreviventes, mortos e gênero
plt.figure(figsize=(8, 6))
sns.countplot(x='sex', hue='survived', data=titanic_df)
plt.title('Relação entre Sobreviventes e Mortos por Gênero no Titanic')
plt.xlabel('Gênero')
plt.ylabel('Número de Pessoas')
plt.xticks(ticks=[0, 1], labels=['Masculino', 'Feminino'])
plt.legend(title='Sobreviveu', labels=['Não', 'Sim'])
plt.show()
```

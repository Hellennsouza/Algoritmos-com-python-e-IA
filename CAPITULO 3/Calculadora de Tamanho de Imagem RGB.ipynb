```python
#Enunciado
#Crie um notebook no Colab que calcule o tamanho bruto de uma imagem RGB. O notebook
#deve permitir alterar largura e altura manualmente no código, e mostrar o resultado em
#bytes, KB e MB.

A = 10 #Substitua o 10 por  a Altura da Imagem
B = 10 #Substitua o 10 por  a Digite a Largura da Imagem


Bytes = A*B*3
KB = Bytes/1024
MegaBytes = (Bytes / 1024) / 1024

print("Bytes = ", Bytes)
print("KB = ", KB)
print("MB =", MegaBytes)
```

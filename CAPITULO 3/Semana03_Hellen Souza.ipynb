### 🧠 O que aprendi sobre bits

Bits são a menor unidade de informação em um computador, representando um 0 ou um 1. Eles são os blocos de construção fundamentais de toda a informação digital, desde textos e imagens até áudio e vídeo. A combinação de bits forma bytes (geralmente 8 bits), que permitem representar uma gama maior de valores e caracteres.

Compreender bits é crucial para entender como os dados são armazenados, processados e transmitidos. Por exemplo, a profundidade de cor de uma imagem é determinada pelo número de bits usados para representar cada pixel. Da mesma forma, a qualidade de um arquivo de áudio depende da taxa de amostragem e da profundidade de bits, que influenciam a precisão com que o som original é digitalizado.

```python
# Cálculo 1: Conversão de Decimal para Binário e vice-versa

numero_decimal = 25
numero_binario = bin(numero_decimal)  # Converte para binário (prefixo '0b')

print(f"O número decimal {numero_decimal} em binário é: {numero_binario}")

# Convertendo de volta para decimal
numero_decimal_revertido = int(numero_binario, 2) # O '2' indica base binária
print(f"O número binário {numero_binario} em decimal é: {numero_decimal_revertido}")
```

```python
# Cálculo 2: Representação de cores usando bits (RGB)

# Cada componente (Red, Green, Blue) é geralmente representado por 8 bits
# O valor 255 em decimal é 11111111 em binário (8 bits)

vermelho_decimal = 255 # Intensidade máxima de vermelho
verde_decimal = 128  # Média intensidade de verde
azul_decimal = 0    # Sem azul

print(f"Componente Vermelho (Decimal): {vermelho_decimal}, (Binário): {bin(vermelho_decimal)}")
print(f"Componente Verde (Decimal): {verde_decimal}, (Binário): {bin(verde_decimal)}")
print(f"Componente Azul (Decimal): {azul_decimal}, (Binário): {bin(azul_decimal)}")

```

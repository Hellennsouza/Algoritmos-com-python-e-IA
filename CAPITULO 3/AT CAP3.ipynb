Use o prompt abaixo para pedir a uma IA que explique a conversão de 42 para binário.

**PROMPT:**


Explique para um estudante iniciante como converter o número
decimal 42 para binário usando divisões sucessivas por 2.
Mostre cada etapa, explique por que os restos são lidos de
baixo para cima e inclua uma verificação somando potências
de 2.

**RETORNO DA IA:**

Para converter o número decimal 42 em binário, dividimos o número sucessivamente pela base que queremos encontrar (2), registrando o quociente e o resto de cada operação até que o quociente seja zero.

O Processo de Divisão
42 ÷ 2 = 21, com resto 0

21 ÷ 2 = 10, com resto 1

10 ÷ 2 = 5, com resto 0

5 ÷ 2 = 2, com resto 1

2 ÷ 2 = 1, com resto 0

1 ÷ 2 = 0, com resto 1

Lendo os restos de baixo para cima, o resultado binário é 101010.

Por que lemos os restos de baixo para cima?

A primeira divisão por 2 determina se o número original é par ou ímpar, o que define a casa de menor valor (o bit menos significativo, posicionado na extrema direita do número binário).

À medida que avançamos nas divisões, estamos desmembrando o número em blocos cada vez maiores que representam potências superiores de 2.A última divisão realizada extrai a maior potência de 2 que cabe dentro do número, tornando-se o bit de maior valor (o bit mais significativo).

Como o nosso sistema de escrita numérica organiza os números do maior valor à esquerda para o menor valor à direita, precisamos posicionar o último resto encontrado na extrema esquerda.

Verificação com Potências de 2

Para confirmar se o número binário 101010 equivale ao decimal 42, multiplicamos cada bit, da direita para a esquerda,
pelas potências crescentes de 2 (começando em $2^0$) e somamos os resultados:

0 × $2^0$ = 0 × 1 = 0

1 × $2^1$ = 1 × 2 = 2

0 × $2^2$ = 0 × 4 = 0

1 × $2^3$ = 1 × 8 = 8

0 × $2^4$ = 0 × 16 = 0

1 × $2^5$ = 1 × 32 = 32

Somando todos os valores não zerados: 32 + 8 + 2 = 42.

A conversão está perfeitamente correta e coincide com o exercício demonstrado na sua anotação no Colab.

```python
#Enunciado
#Peça a uma IA que gere um código Python para calcular o tamanho de uma imagem RGB.
#Em seguida, audite o código usando os cinco critérios abaixo.
#1. Usa largura, altura e 3 bytes por pixel? Resposta: sim
#2. Converte corretamente para KB e MB? Resposta: sim
#3. Usa nomes de variáveis compreensíveis? Resposta: sim
#4. Possui comentários úteis? Resposta: sim
#5. Você consegue explicar sem copiar? Resposta: sim, porém com excessão do ":.2f"



#código retornado pela IA

def calcular_tamanho_imagem_rgb(largura, altura):
    # Em uma imagem RGB sem compressão, cada pixel ocupa 3 bytes (1 byte para Red, 1 para Green, 1 para Blue)
    bytes_por_pixel = 3

    # Cálculo do tamanho bruto em bytes
    tamanho_bytes = largura * altura * bytes_por_pixel

    # Conversão para Kilobytes (KB) e Megabytes (MB)
    tamanho_kb = tamanho_bytes / 1024
    tamanho_mb = tamanho_kb / 1024

    return tamanho_bytes, tamanho_kb, tamanho_mb

# Exemplo de uso para uma imagem Full HD (1920 x 1080)
largura = 1920
altura = 1080

t_bytes, t_kb, t_mb = calcular_tamanho_imagem_rgb(largura, altura)

print(f"Resolução: {largura} x {altura}")
print(f"Tamanho em Bytes: {t_bytes}")
print(f"Tamanho em KB: {t_kb:.2f}")
print(f"Tamanho em MB: {t_mb:.2f}")
```

Implementar uma função que recebe 3 notas e calcula a media

```python

def calcular_media(a,b,c,aprovacao):
  media = (a + b + c) / 3
  if media >= aprovacao:
    resultado = "Aprovado"
  else:
    resultado = "Reprovado"
  return media, resultado
```

```python
r, m = calcular_media(5,7,10,6)
```

```python
def vendas(lista_vendas):
  soma = sum(lista_vendas)
  media = soma / len(lista_vendas)
  maior_venda = max(lista_vendas)
  menor_venda = min(lista_vendas)
  return soma, media, maior_venda
```

```python
v = [456,34.56,459,234.5,89]
s,t,m = vendas(v)
```

```python
print(f"A soma das vendas foi {s}, com media de {t}")
print(f"O maior valor vendido foi {m}")
```

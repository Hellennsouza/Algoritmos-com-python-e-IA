#**7º Enunciado:**
O que há de errado no código abaixo?

```python
def calcular_desconto(valor, percentual):
    desconto = valor * percentual / 100

preco_final = 100 - calcular_desconto(100, 10)
print(preco_final)
```

##7º Resposta:
A função calcula o desconto, mas **não tem `return`**. Toda função sem `return` devolve `None`, então a conta vira `100 - None` e gera `TypeError: unsupported operand type(s) for -: 'int' and 'NoneType'`. A correção é devolver o valor calculado:

````python
#7º Resposta (corrigido):
def calcular_desconto(valor, percentual):
    desconto = valor * percentual / 100
    return desconto

preco_final = 100 - calcular_desconto(100, 10)
print(preco_final)
assert calcular_desconto(100, 10) == 10
````

#**8º Enunciado:**

```python
def mostrar_menu():
```

#**9º Enunciado:**
Corrija o erro de escopo:

```python
def criar_mensagem(nome):
    mensagem = f'Olá, {nome}'
    return mensagem

criar_mensagem('Ana')
print(mensagem)   # NameError aqui
```

##9º Resposta:
`mensagem` é uma variável **local**: só existe enquanto a função executa. A função retorna o valor corretamente, mas a chamada `criar_mensagem('Ana')` descarta o resultado. Fora da função, o nome `mensagem` não existe, daí o `NameError`. A correção é guardar o retorno em uma variável do programa principal:

````python
#9º Resposta (corrigido):
def criar_mensagem(nome):
    mensagem = f'Olá, {nome}'
    return mensagem

resultado = criar_mensagem('Ana')
print(resultado)
````

#**11º Enunciado:**
Refatore um programa com dados de alunos para usar quatro funções: calcular_media, classificar_aluno, criar_aluno e mostrar_relatorio. O código principal deve ter no máximo dez linhas.

````python
#11º Resposta:
def calcular_media(n1, n2, n3):
    """Retorna a média aritmética de três notas."""
    return (n1 + n2 + n3) / 3


def classificar_aluno(media):
    """Retorna a situação do aluno pela média."""
    if media >= 7:
        return 'Aprovado'
    elif media >= 5:
        return 'Recuperação'
    return 'Reprovado'


def criar_aluno(nome, n1, n2, n3):
    """Monta o dicionário do aluno reaproveitando as outras funções."""
    media = calcular_media(n1, n2, n3)
    return {'nome': nome, 'n1': n1, 'n2': n2, 'n3': n3,
            'media': media, 'situacao': classificar_aluno(media)}


def mostrar_relatorio(aluno):
    """Única função que imprime."""
    print(f"Aluno   : {aluno['nome']}")
    print(f"Notas   : {aluno['n1']} | {aluno['n2']} | {aluno['n3']}")
    print(f"Média   : {aluno['media']:.2f}")
    print(f"Situação: {aluno['situacao']}")
    print('-' * 30)


# Testes das funções de cálculo
assert calcular_media(10, 8, 6) == 8.0
assert classificar_aluno(7.0) == 'Aprovado'
assert classificar_aluno(5.0) == 'Recuperação'
assert classificar_aluno(4.99) == 'Reprovado'
assert criar_aluno('Teste', 10, 8, 6)['situacao'] == 'Aprovado'

# ---------- Código principal (4 linhas) ----------
turma = [criar_aluno('Ana', 8.0, 7.5, 9.0), criar_aluno('Bruno', 6.5, 5.0, 6.0), criar_aluno('Carla', 4.0, 5.5, 3.0)]
for aluno in turma:
    mostrar_relatorio(aluno)
print('Testes passaram!')
````

#**12º Enunciado:**
Crie estatisticas_vendas(vendas) que retorne três valores: total vendido, menor venda e maior venda. Se a lista estiver vazia, retornar 0.0, 0.0, 0.0.

````python
#12º Resposta:
def estatisticas_vendas(vendas):
    """Retorna (total, menor, maior) de uma lista de vendas."""
    if len(vendas) == 0:            # min([]) e max([]) gerariam ValueError
        return 0.0, 0.0, 0.0
    return sum(vendas), min(vendas), max(vendas)

assert estatisticas_vendas([]) == (0.0, 0.0, 0.0)
assert estatisticas_vendas([100]) == (100, 100, 100)
assert estatisticas_vendas([100, 250, 50]) == (400, 50, 250)

total, menor, maior = estatisticas_vendas([1200.50, 980.00, 1350.75, 1100.00, 1580.90])
print(f'Total: R$ {total:.2f} | Menor: R$ {menor:.2f} | Maior: R$ {maior:.2f}')
````

#**13º Enunciado:**
Implemente ordenar_alunos(turma, criterio): ordena por "nome" (crescente) ou por "media" (decrescente), sem modificar a lista original. Critério inválido retorna nova lista igual à original.

````python
#13º Resposta:
def ordenar_alunos(turma, criterio):
    """Retorna uma NOVA lista ordenada; a turma original não muda."""
    if criterio == 'nome':
        return sorted(turma, key=lambda aluno: aluno['nome'])
    elif criterio == 'media':
        return sorted(turma, key=lambda aluno: aluno['media'], reverse=True)
    return list(turma)              # critério inválido: cópia na mesma ordem


alunos = [
    {'nome': 'Carla', 'media': 9.1},
    {'nome': 'Ana', 'media': 8.5},
    {'nome': 'Bruno', 'media': 5.8},
]

por_nome = ordenar_alunos(alunos, 'nome')
por_media = ordenar_alunos(alunos, 'media')
invalido = ordenar_alunos(alunos, 'idade')

assert [a['nome'] for a in por_nome] == ['Ana', 'Bruno', 'Carla']
assert [a['nome'] for a in por_media] == ['Carla', 'Ana', 'Bruno']
assert invalido == alunos and invalido is not alunos     # igual, mas é outra lista
assert [a['nome'] for a in alunos] == ['Carla', 'Ana', 'Bruno']   # original intacta

for a in por_media:
    print(f"{a['nome']}: {a['media']}")
print('Testes passaram!')
````

##14º Resposta — Prompt Contract
```
PROMPT CONTRACT
Objetivo:    Calcular a taxa de conversão em porcentagem (proporção de
             visitantes que realizaram uma compra).
Parâmetros:  visitas: int ou float — número de visitantes.
             compras: int ou float — número de compras realizadas.
Retorno:     float — taxa em porcentagem (5.0 para 5%, não 0.05).
Restrições:  - Se visitas for zero, retornar 0.0 (sem dividir por zero).
```

````python
#14º Resposta — minha implementação:
def calcular_taxa_conversao(visitas, compras):
    """Retorna a taxa de conversão em porcentagem."""
    if visitas == 0:
        return 0.0
    if compras > visitas:
        return 0.0
    return compras / visitas * 100

assert calcular_taxa_conversao(200, 10) == 5.0
assert calcular_taxa_conversao(0, 5) == 0.0         # sem visitas
assert calcular_taxa_conversao(10, 20) == 0.0       # compras > visitas
assert calcular_taxa_conversao(50, 0) == 0.0
print('Minha versão: testes passaram!')
````

````python
#14º Resposta — versão gerada pela IA (Claude), testada com os MESMOS asserts:
def calcular_taxa_conversao_ia(visitas, compras):
    if visitas <= 0:
        return 0.0
    return round((compras / visitas) * 100, 2)

testes = [((200, 10), 5.0), ((0, 5), 0.0), ((10, 20), 0.0), ((50, 0), 0.0)]
for (v, c), esperado in testes:
    obtido = calcular_taxa_conversao_ia(v, c)
    status = 'ok' if obtido == esperado else 'FALHOU'
    print(f'visitas={v:>3} compras={c:>2} -> esperado {esperado:>5} | obtido {obtido:>6}  {status}')
````

| Compras > visitas | Retorna 0.0, como o contrato pede | **Retorna 200.0**: ignorou essa restrição do contrato |
| Formato do retorno | Porcentagem sem arredondar | Porcentagem arredondada com `round(..., 2)` |

**Conclusão:** a IA acertou o caso principal e o de divisão por zero, mas descumpriu a regra de compras > visitas, o que só apareceu porque escrevi o teste antes de aceitar o código.
````python
#15º Resposta — função gerada pela IA (Claude), ainda NÃO executada nos testes:
def estatisticas_vendas_ia(vendas):
    """Calcula total, menor e maior venda."""
    if not vendas:
        return 0.0, 0.0, 0.0
    vendas.sort()
    return sum(vendas), vendas[0], vendas[-1]
````

````python
#15º Resposta — testes com assert:
assert estatisticas_vendas_ia([]) == (0.0, 0.0, 0.0)
assert estatisticas_vendas_ia([100, 250, 50]) == (400, 50, 250)

vendas_dia = [300, 100, 200]
estatisticas_vendas_ia(vendas_dia)
print('Lista depois da função da IA:', vendas_dia)
print('Lista foi preservada?', vendas_dia == [300, 100, 200])

# versão corrigida: usa min() e max(), sem alterar a lista
def estatisticas_vendas_corrigida(vendas):
    if len(vendas) == 0:
        return 0.0, 0.0, 0.0
    return sum(vendas), min(vendas), max(vendas)

vendas_dia = [300, 100, 200]
assert estatisticas_vendas_corrigida(vendas_dia) == (600, 100, 300)
assert vendas_dia == [300, 100, 200]        # original intacta
assert estatisticas_vendas_corrigida([]) == (0.0, 0.0, 0.0)
print('Versão corrigida: testes passaram e a lista original foi preservada!')
````

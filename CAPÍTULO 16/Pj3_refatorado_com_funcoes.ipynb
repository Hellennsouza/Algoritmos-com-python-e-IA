#**Projeto do capítulo — Pj3 refatorado com funções**
Reorganize o Projeto 3 usando funções. O comportamento deve ser o mesmo, mas o código deve ficar mais claro, modular e testável.

```python
#Projeto — Resposta (funções):
import math


def ler_nota(mensagem):
    """Lê uma nota e repete até ela estar entre 0 e 10."""
    nota = float(input(mensagem))
    while nota < 0 or nota > 10:
        nota = float(input('  Nota inválida (0 a 10). ' + mensagem.strip() + ' '))
    return nota


def calcular_media(n1, n2, n3, pesos=(0.3, 0.3, 0.4)):
    """Média ponderada de três notas; os pesos têm valor padrão."""
    return n1 * pesos[0] + n2 * pesos[1] + n3 * pesos[2]


def classificar_aluno(media, aprov=7.0, recup=5.0):
    """Retorna Aprovado, Recuperação ou Reprovado."""
    if media >= aprov:
        return 'Aprovado'
    elif media >= recup:
        return 'Recuperação'
    return 'Reprovado'


def criar_aluno(nome, p1, p2, p3):
    """Monta o dicionário do aluno com média e situação."""
    media = calcular_media(p1, p2, p3)
    return {'nome': nome, 'p1': p1, 'p2': p2, 'p3': p3,
            'media': media, 'situacao': classificar_aluno(media)}


def cadastrar_turma():
    """Pergunta a quantidade de alunos e lê nome e notas de cada um."""
    n = int(input('Quantos alunos? '))
    while n <= 0:
        n = int(input('A quantidade deve ser maior que zero. Quantos alunos? '))
    turma = []
    for i in range(n):
        print(f'Aluno {i + 1} de {n}')
        nome = input('  Nome: ').strip().title()
        turma.append(criar_aluno(nome, ler_nota('  P1: '), ler_nota('  P2: '), ler_nota('  P3: ')))
    return turma


def estatisticas_turma(turma):
    """Dicionário com média geral, mínima, máxima, desvio padrão e contagens."""
    stats = {'media_geral': 0.0, 'minima': 0.0, 'maxima': 0.0, 'desvio_pad': 0.0,
             'Aprovado': 0, 'Recuperação': 0, 'Reprovado': 0}
    if len(turma) == 0:                 # guarda: lista vazia
        return stats

    medias = []
    for aluno in turma:
        medias.append(aluno['media'])
        stats[aluno['situacao']] += 1

    media_geral = sum(medias) / len(medias)
    soma_quadrados = 0
    for m in medias:
        soma_quadrados += (m - media_geral) ** 2

    stats['media_geral'] = media_geral
    stats['minima'] = min(medias)
    stats['maxima'] = max(medias)
    stats['desvio_pad'] = math.sqrt(soma_quadrados / len(medias))
    return stats


def ranking_turma(turma):
    """Nova lista da maior para a menor média (não altera a turma)."""
    return sorted(turma, key=lambda a: a['media'], reverse=True)


def buscar_aluno(turma, nome):
    """Retorna (posição no ranking, aluno) ou (0, None) se não encontrar."""
    nome = nome.strip().lower()
    posicao = 1
    for aluno in ranking_turma(turma):
        if aluno['nome'].lower() == nome:
            return posicao, aluno
        posicao += 1
    return 0, None


def mostrar_relatorio(turma):
    """Única função que imprime o relatório completo."""
    print('\nRELATÓRIO DA TURMA (ranking)')
    print('-' * 44)
    posicao = 1
    for aluno in ranking_turma(turma):
        print(f"{posicao}º {aluno['nome']:<12} {aluno['media']:>5.2f}  {aluno['situacao']}")
        posicao += 1
    stats = estatisticas_turma(turma)
    print('-' * 44)
    print(f"Média geral  : {stats['media_geral']:.2f}")
    print(f"Menor média  : {stats['minima']:.2f}")
    print(f"Maior média  : {stats['maxima']:.2f}")
    print(f"Desvio padrão: {stats['desvio_pad']:.2f}")
    print(f"Aprovados    : {stats['Aprovado']}")
    print(f"Recuperação  : {stats['Recuperação']}")
    print(f"Reprovados   : {stats['Reprovado']}")
```

```python
#Projeto — Testes com assert:
assert calcular_media(7, 7, 7) == 7.0
assert round(calcular_media(10, 8, 6), 2) == 7.8      # 3.0 + 2.4 + 2.4
assert round(calcular_media(9, 9, 8), 2) == 8.6
assert calcular_media(10, 8, 6, pesos=(1/3, 1/3, 1/3)) == 8.0

assert classificar_aluno(7.0) == 'Aprovado'
assert classificar_aluno(5.0) == 'Recuperação'
assert classificar_aluno(4.99) == 'Reprovado'
assert classificar_aluno(6.0, aprov=6.0) == 'Aprovado'

aluno_t = criar_aluno('Teste', 9, 9, 8)
assert aluno_t['situacao'] == 'Aprovado'

turma_t = [criar_aluno('A', 5, 5, 5), criar_aluno('B', 9, 9, 8)]
assert ranking_turma(turma_t)[0]['nome'] == 'B'
assert turma_t[0]['nome'] == 'A'                       # ranking não alterou a turma
assert buscar_aluno(turma_t, '  b ')[0] == 1
assert buscar_aluno(turma_t, 'Z') == (0, None)

assert estatisticas_turma([])['media_geral'] == 0.0     # lista vazia não quebra
assert estatisticas_turma(turma_t)['Aprovado'] == 1
print('Todos os testes passaram!')
```

```python
#Projeto — Código principal:
turma = cadastrar_turma()
mostrar_relatorio(turma)
posicao, aluno = buscar_aluno(turma, input('\nBuscar aluno: '))
if aluno is None:
    print('Aluno não encontrado.')
else:
    print(f"{aluno['nome']}: {aluno['media']:.2f} - {aluno['situacao']} ({posicao}º lugar)")
```

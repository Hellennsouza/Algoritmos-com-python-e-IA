#**3º Enunciado:**

Complete as condições para três situações:

verificar se idade é maior ou igual a 18;

verificar se senha é igual a "abc123";

verificar se total é menor ou igual a orcamento.

```python
#3º Resposta:
idade = int(input("Digite sua idade: "))
senha = input("Digite sua senha: ")
total = float(input("Digite o valor total da compra: "))
orcamento = float(input("Digite o valor do orçamento: "))

if idade >= 18:
  print(True)
else:
  print(False)


if senha == "abc123":
  print(True)
else:
  print(False)

if total <= orcamento:
  print(True)
else:
  print(False)
```

#**4º Enunciado:**

Crie um programa que leia uma temperatura e a classifique:

abaixo de 15°C é frio;

de 15°C até (sem incluir) 25°C é agradável;

25°C ou mais é quente.

```python
#4º Resposta:
temperatura = float(input("Digite a temperatura: "))

if temperatura < 15:
  print("Frio")
elif temperatura >= 15 and temperatura < 25:
  print("Quente")
else:
  print("Muito quente")
```

#**5º Enunciado:**

Leia o valor de uma compra. Se o valor for maior ou igual a R$ 200,00, aplique 10% de
desconto. Caso contrário, mantenha o valor original. Exiba o valor final.

```python
valor_compra = float(input("Digite o valor da compra: "))

if valor_compra >= 200:
  valor_final = valor_compra * 0.9
else:
  valor_final = valor_compra
print(f"O valor final é R$ {valor_final:.2f}")
```

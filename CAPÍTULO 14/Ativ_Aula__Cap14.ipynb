```python
notas = []
discentes = []

for i in range(10):
    nome = input(f"Digite o nome do discente {i+1}: ")
    discentes.append(nome)

    nota = float(input(f"Digite a nota {i+1}: "))
    notas.append(nota)

for i in range(10):
    print(f'{discentes[i]} teve a nota = {notas[i]}')

media = sum(notas) / len(notas)
print(f"A média das notas é: {media:.2f}")
```

```python
notas = []

while True:
  nota = float(input('Nota(-1 para sair):'))
  if nota == -1:
    break
  notas.append(nota)
if len(notas)==0:
    print('Nenhuma nota foi digitada')
else:

  media = sum(notas) / len(notas)
  print(f"A média das notas é: {media:.2f}")
```

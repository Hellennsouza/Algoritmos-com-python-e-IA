#**6º Enunciado:**
Crie uma lista com seis produtos. Cada produto deve ser um dicionário com nome, categoria e preco. Calcule: produto mais caro; produto mais barato; preço médio; produtos com preço acima da média; lista ordenada por preço crescente.

````python
#6º Resposta:
produtos = [
    {'nome': 'Notebook', 'categoria': 'Informática', 'preco': 3500.00},
    {'nome': 'Mouse', 'categoria': 'Informática', 'preco': 89.90},
    {'nome': 'Cadeira', 'categoria': 'Móveis', 'preco': 750.00},
    {'nome': 'Caderno', 'categoria': 'Papelaria', 'preco': 25.50},
    {'nome': 'Monitor', 'categoria': 'Informática', 'preco': 1200.00},
    {'nome': 'Luminária', 'categoria': 'Móveis', 'preco': 140.00},
]

mais_caro = max(produtos, key=lambda p: p['preco'])
mais_barato = min(produtos, key=lambda p: p['preco'])

soma = 0
for p in produtos:
    soma += p['preco']
preco_medio = soma / len(produtos)

print(f"Mais caro: {mais_caro['nome']} (R$ {mais_caro['preco']:.2f})")
print(f"Mais barato: {mais_barato['nome']} (R$ {mais_barato['preco']:.2f})")
print(f'Preço médio: R$ {preco_medio:.2f}')

print('\nAcima da média:')
for p in produtos:
    if p['preco'] > preco_medio:
        print(f"  - {p['nome']} (R$ {p['preco']:.2f})")

print('\nOrdenados por preço (crescente):')
for p in sorted(produtos, key=lambda p: p['preco']):
    print(f"  {p['nome']:<10} R$ {p['preco']:>8.2f}")
````

#**7º Enunciado:**
Explique por que o código abaixo gera erro e corrija-o para exibir o último nome da lista.

```python
nomes = ["Ana", "Bruno", "Carla"]
print(nomes[3])
```

##7º Resposta:
Os índices começam em 0. A lista tem 3 elementos, então os índices válidos são 0, 1 e 2. `nomes[3]` pede um quarto elemento que não existe e gera `IndexError: list index out of range`. A correção é usar o índice 2 ou, melhor ainda, `-1`, que sempre aponta para o último elemento, qualquer que seja o tamanho da lista.

````python
#7º Resposta (corrigido):
nomes = ['Ana', 'Bruno', 'Carla']
print(nomes[2])     # índice correto do último
print(nomes[-1])    # índice negativo: funciona para qualquer tamanho
````

#**8º Enunciado:**
Analise o código abaixo. Explique o erro e proponha uma correção usando if.

```python
notas = []
media = sum(notas) / len(notas)
print(media)
```

##8º Resposta:
A lista está vazia: `sum(notas)` dá 0 (isso não é erro), mas `len(notas)` também dá 0, e a divisão `0 / 0` gera `ZeroDivisionError: division by zero`. A correção é verificar se a lista tem elementos antes de dividir:

````python
#8º Resposta (corrigido):
notas = []

if len(notas) > 0:
    media = sum(notas) / len(notas)
    print(f'Média: {media:.2f}')
else:
    print('Não há notas cadastradas para calcular a média.')
````

#**9º Enunciado:**
Explique o que faz o código abaixo e escreva a saída esperada ao exibir o ranking.

```python
alunos = [
    {"nome": "Ana", "media": 8.5},
    {"nome": "Bruno", "media": 6.0},
    {"nome": "Carla", "media": 9.0}
]
ranking = sorted(alunos, key=lambda aluno: aluno["media"], reverse=True)
```

##9º Resposta:
| Trecho | O que faz |
|---|---|
| `sorted(alunos, ...)` | Cria uma **nova** lista ordenada; `alunos` continua na ordem original |
| `key=lambda aluno: aluno["media"]` | Critério de ordenação: para cada dicionário, usa o valor da chave `media` |
| `reverse=True` | Ordena do maior para o menor |

Saída esperada: **1º Carla – 9.0, 2º Ana – 8.5, 3º Bruno – 6.0**.

````python
#9º Resposta (conferindo):
alunos = [
    {'nome': 'Ana', 'media': 8.5},
    {'nome': 'Bruno', 'media': 6.0},
    {'nome': 'Carla', 'media': 9.0}
]
ranking = sorted(alunos, key=lambda aluno: aluno['media'], reverse=True)

for posicao, aluno in enumerate(ranking, start=1):
    print(f"{posicao}º {aluno['nome']} - {aluno['media']:.1f}")
````

#**10º Enunciado:**
Implemente o Projeto 3 completo com cadastro de alunos, cálculo de média, classificação, relatório individual, ranking e estatísticas gerais. Adicione também: validação de notas entre 0 e 10; busca de aluno pelo nome; exibição de alunos agrupados por situação.

````python
#10º Resposta:
# ---------- Cadastro ----------
n_alunos = int(input('Quantos alunos? '))
while n_alunos <= 0:
    n_alunos = int(input('A quantidade deve ser maior que zero. Quantos alunos? '))

turma = []
for i in range(n_alunos):
    print(f'\nAluno {i + 1} de {n_alunos}')
    nome = input('  Nome: ').strip().title()

    notas = []
    for n in range(1, 4):
        nota = float(input(f'  Nota {n}: '))
        while nota < 0 or nota > 10:
            nota = float(input(f'  Nota inválida (0 a 10). Nota {n}: '))
        notas.append(nota)

    media = sum(notas) / len(notas)
    if media >= 7:
        situacao = 'Aprovado'
    elif media >= 5:
        situacao = 'Recuperação'
    else:
        situacao = 'Reprovado'

    turma.append({'nome': nome, 'notas': notas, 'media': media, 'situacao': situacao})

# ---------- Relatório individual ----------
print('\n===== RELATÓRIO INDIVIDUAL =====')
print(f"{'Nome':<15}{'N1':>5}{'N2':>5}{'N3':>5}{'Média':>8}  Situação")
for aluno in turma:
    n1, n2, n3 = aluno['notas']
    print(f"{aluno['nome']:<15}{n1:>5.1f}{n2:>5.1f}{n3:>5.1f}{aluno['media']:>8.2f}  {aluno['situacao']}")

# ---------- Ranking ----------
ranking = sorted(turma, key=lambda a: a['media'], reverse=True)
print('\n===== RANKING =====')
for posicao, aluno in enumerate(ranking, start=1):
    print(f"{posicao}º {aluno['nome']} - {aluno['media']:.2f}")

# ---------- Estatísticas gerais ----------
medias = []
for aluno in turma:
    medias.append(aluno['media'])
media_geral = sum(medias) / len(medias)

aprovados = 0
recuperacao = 0
reprovados = 0
for aluno in turma:
    if aluno['situacao'] == 'Aprovado':
        aprovados += 1
    elif aluno['situacao'] == 'Recuperação':
        recuperacao += 1
    else:
        reprovados += 1

print('\n===== ESTATÍSTICAS =====')
print(f'Média geral: {media_geral:.2f}')
print(f'Maior média: {max(medias):.2f} | Menor média: {min(medias):.2f}')
print(f'Aprovados: {aprovados} ({aprovados / n_alunos:.0%})')
print(f'Em recuperação: {recuperacao} ({recuperacao / n_alunos:.0%})')
print(f'Reprovados: {reprovados} ({reprovados / n_alunos:.0%})')

# ---------- Agrupados por situação ----------
print('\n===== POR SITUAÇÃO =====')
for grupo in ['Aprovado', 'Recuperação', 'Reprovado']:
    print(f'{grupo}:')
    encontrou = False
    for aluno in turma:
        if aluno['situacao'] == grupo:
            print(f"  - {aluno['nome']} ({aluno['media']:.2f})")
            encontrou = True
    if not encontrou:
        print('  (nenhum)')

# ---------- Busca pelo nome ----------
busca = input('\nBuscar aluno pelo nome: ').strip().lower()
achado = False
for aluno in turma:
    if aluno['nome'].lower() == busca:
        print(f"{aluno['nome']}: média {aluno['media']:.2f} - {aluno['situacao']}")
        achado = True
if not achado:
    print('Aluno não encontrado.')
````

#**14º Enunciado:**
Peça à IA duas versões para encontrar o aluno com maior média: uma usando for; outra usando max() com key=lambda. Compare as duas.

````python
#14º Resposta:
# Versão 1 — com for (mais explícita)
melhor = turma[0]
for aluno in turma:
    if aluno['media'] > melhor['media']:
        melhor = aluno
print(f"Versão for: {melhor['nome']} - {melhor['media']:.2f}")

# Versão 2 — com max() e lambda (mais compacta)
melhor = max(turma, key=lambda aluno: aluno['media'])
print(f"Versão max: {melhor['nome']} - {melhor['media']:.2f}")
````

##14º Resposta (comparação):
| Critério | Versão com `for` | Versão com `max()` + `lambda` |
|---|---|---|
| Resultado | Ana Silva – 8.17 | Ana Silva – 8.17 (o mesmo) |
| Linhas | 4 | 1 |
| Legibilidade para iniciante | Cada passo da comparação aparece | Exige entender `key=lambda` |
| Empate | Fica com o primeiro (usa `>`) | Também fica com o primeiro |
| Lista vazia | `turma[0]` gera `IndexError` | `max()` gera `ValueError` |
| Facilidade de adaptar | Fácil incluir regras extras dentro do `if` | Menos flexível para regras compostas |

**Decisão:** uso `max()` com `lambda` no código final por ser direto e menos sujeito a erro de digitação, mas entendo a versão com `for`, que mostra o que o `max()` faz por dentro. Nas duas, protejo a chamada com `if len(turma) > 0`.

#**15º Enunciado:**
Escreva uma célula Markdown explicando como Pandas simplificaria o Projeto 3. Depois, pesquise e escreva uma versão exploratória usando pd.DataFrame(), mean(), sort_values() e describe().

##15º Resposta — Como o Pandas simplificaria o Projeto 3
No Projeto 3, a turma é uma **lista de dicionários**: cada dicionário é uma linha e cada chave é uma coluna. O Pandas oferece o **DataFrame**, uma tabela pronta em que essas operações viram uma linha cada:

- **média geral:** em vez de um laço acumulando médias, `df['media'].mean()`;
- **ranking:** em vez de `sorted(..., key=lambda ...)`, `df.sort_values('media', ascending=False)`;
- **contagem por situação:** em vez de três contadores, `df['situacao'].value_counts()`;
- **estatísticas gerais:** `df.describe()` mostra quantidade, média, desvio padrão, mínimo, quartis e máximo de uma vez.

Entender listas e dicionários continua essencial: quando escrevemos `df['media'].mean()`, o Pandas faz exatamente o que fizemos à mão (somar e dividir pela quantidade).

````python
#15º Resposta — versão exploratória com Pandas:
import pandas as pd

df = pd.DataFrame(turma)
df[['nota1', 'nota2', 'nota3']] = pd.DataFrame(df['notas'].tolist())
df = df.drop(columns='notas')

print('Tabela:')
print(df.round(2).to_string(index=False))

print(f"\nMédia geral: {df['media'].mean():.2f}")

print('\nRanking:')
print(df.sort_values('media', ascending=False)[['nome', 'media', 'situacao']].round(2).to_string(index=False))

print('\nAlunos por situação:')
print(df['situacao'].value_counts().to_string())

print('\nEstatísticas descritivas:')
print(df[['nota1', 'nota2', 'nota3', 'media']].describe().round(2))
````

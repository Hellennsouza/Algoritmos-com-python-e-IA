#**Projeto do capítulo — Projeto 3: Sistema de análise de turma**
Versão final do Projeto 3, reunindo em um único programa o que foi pedido nos exercícios 10, 11 e 12: cadastro com validação de notas, média e classificação, relatório individual, ranking, estatísticas gerais, agrupamento por situação, interpretação da média geral e busca de aluno com posição no ranking.

```python
#Projeto — Resposta:
# ===== 1. Cadastro =====
n_alunos = int(input('Quantos alunos? '))
while n_alunos <= 0:
    n_alunos = int(input('A quantidade deve ser maior que zero. Quantos alunos? '))

turma = []
for i in range(n_alunos):
    print(f'\nAluno {i + 1} de {n_alunos}')
    nome = input('  Nome: ').strip().title()
    while nome == '':
        nome = input('  O nome não pode ficar vazio. Nome: ').strip().title()

    notas = []
    for n in range(1, 4):
        nota = float(input(f'  Nota {n}: '))
        while nota < 0 or nota > 10:
            nota = float(input(f'  Nota inválida (0 a 10). Nota {n}: '))
        notas.append(nota)

    media = sum(notas) / len(notas)
    if media >= 7:
        situacao = 'Aprovado'
    elif media >= 5:
        situacao = 'Recuperação'
    else:
        situacao = 'Reprovado'

    turma.append({'nome': nome, 'nota1': notas[0], 'nota2': notas[1], 'nota3': notas[2],
                  'media': media, 'situacao': situacao})

# ===== 2. Relatório individual (ordem de cadastro) =====
print('\n' + '=' * 55)
print('RELATÓRIO INDIVIDUAL')
print('=' * 55)
print(f"{'Nome':<15}{'N1':>5}{'N2':>5}{'N3':>5}{'Média':>8}  Situação")
for a in turma:
    print(f"{a['nome']:<15}{a['nota1']:>5.1f}{a['nota2']:>5.1f}{a['nota3']:>5.1f}{a['media']:>8.2f}  {a['situacao']}")

# ===== 3. Ranking (sorted preserva a lista turma) =====
ranking = sorted(turma, key=lambda a: a['media'], reverse=True)
print('\nRANKING')
for posicao, a in enumerate(ranking, start=1):
    print(f"{posicao}º {a['nome']:<15} {a['media']:.2f}")

# ===== 4. Estatísticas gerais =====
medias = []
contagem = {'Aprovado': 0, 'Recuperação': 0, 'Reprovado': 0}
for a in turma:
    medias.append(a['media'])
    contagem[a['situacao']] += 1

media_geral = sum(medias) / len(medias)
print('\nESTATÍSTICAS')
print(f'Média geral: {media_geral:.2f}')
print(f"Maior média: {max(medias):.2f} ({ranking[0]['nome']})")
print(f"Menor média: {min(medias):.2f} ({ranking[-1]['nome']})")
for situacao in contagem:
    qtd = contagem[situacao]
    print(f'{situacao}: {qtd} ({qtd / n_alunos:.0%})')

# ===== 5. Interpretação =====
if media_geral >= 7:
    print('Interpretação: Turma com bom desempenho geral')
elif media_geral >= 5:
    print('Interpretação: Turma com desempenho intermediário')
else:
    print('Interpretação: Turma com desempenho crítico')

# ===== 6. Agrupados por situação =====
print('\nALUNOS POR SITUAÇÃO')
for grupo in ['Aprovado', 'Recuperação', 'Reprovado']:
    nomes = []
    for a in turma:
        if a['situacao'] == grupo:
            nomes.append(a['nome'])
    if len(nomes) > 0:
        print(f"{grupo}: {', '.join(nomes)}")
    else:
        print(f'{grupo}: (nenhum)')

# ===== 7. Busca com posição no ranking =====
busca = input('\nBuscar aluno pelo nome: ').strip().lower()
encontrado = False
for posicao, a in enumerate(ranking, start=1):
    if a['nome'].lower() == busca:
        print(f"{a['nome']}: média {a['media']:.2f} | {a['situacao']} | {posicao}º de {len(ranking)}")
        encontrado = True
if not encontrado:
    print('Aluno não encontrado.')
```

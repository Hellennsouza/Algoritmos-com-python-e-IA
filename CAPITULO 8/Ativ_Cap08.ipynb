#**1º Enunciado:**
Peça dois números ao usuário e exiba os resultados das quatro operações aritméticas básicas entre eles: soma, subtração, multiplicação e divisão, cada uma com uma etiqueta
clara.

```python
#1º Resposta:
num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))
print("Soma: ", num1 + num2)
print("Subtração: ", num1 - num2)
print("Multiplicação: ", num1 * num2)
print("Divisão: ", num1 / num2)
```

#**2º Enunciado:**
Peça a quantidade total de produtos e a quantidade de produtos que cabem em cada caixa,
e exiba quantas caixas completas podem ser montadas e quantos produtos restam sem
formar uma caixa completa.

```python
#2º Resposta:
total_produtos = int(input("Digite a quantidade total de produtos: "))
produtos_por_caixa = int(input("Digite a quantidade de produtos que cabem em cada caixa: "))
caixas_completas = total_produtos // produtos_por_caixa
produtos_restantes = total_produtos % produtos_por_caixa
print("Caixas completas: ", caixas_completas)
print("Produtos restantes: ", produtos_restantes)
```

#**3º Enunciado:**

Peça um número inteiro e exiba uma mensagem informando se ele é par ou ímpar.

```python
#3º Resposta:
num = int(input("Digite um número inteiro: "))
if num % 2 == 0:
  print("O número é par")
else:
  print("O número é ímpar")
```

#**4º Enunciado:** — Calculadora de desconto

Peça o preço original de um produto e o percentual de desconto. Antes de calcular qualquer
coisa, valide se o preço é maior que zero e se o desconto está entre 0% e 50%; só então
calcule e exiba o valor do desconto e o preço final.

```python
#4º Resposta:
preco_original = float(input("Digite o preço original do produto: "))
percentual_desconto = float(input("Digite o percentual de desconto: "))
if preco_original > 0 and percentual_desconto >= 0 and percentual_desconto <= 50:
  valor_desconto = preco_original * percentual_desconto / 100
  preco_final = preco_original - valor_desconto
  print("Valor do desconto: ", valor_desconto)
  print("Preço final: ", preco_final)
else:
  print("Valores inválidos")
```

#**5º Enunciado:** — Comissão de vendas

Peça o total vendido por um vendedor no mês e calcule a comissão, equivalente a 4% das
vendas. Se o total vendido for maior que R$ 15.000, exiba também uma mensagem de
desempenho excelente.

```python
#5º Resposta:
total_vendas = float(input("Digite o total vendido pelo vendedor no mês: "))
comissao = total_vendas * 0.04
if total_vendas > 15000:
  print("Desempenho excelente!")
print(f"Comissão: R$ {comissao:.2f}")
```

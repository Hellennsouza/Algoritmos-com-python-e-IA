A solução do projeto integra tudo o que foi estudado neste capítulo: operadores aritméticos
para os cálculos, um operador relacional para validar a entrada, e a combinação de um
operador relacional com um operador lógico para decidir o bônus. Para que você possa
comparar com a resposta da IA de forma produtiva, apresentamos aqui uma versão
completa, com comentários que explicam cada bloco de código.

```python

vendas = float(input('Total de vendas no mês (R$): '))
anos_empresa = int(input('Anos de empresa: '))


if vendas < 0 or anos_empresa < 0:
    print('Erro: vendas e anos de empresa não podem ser negativos.')
else:

    salario_base = 2000.0
    comissao = vendas * 0.05

    if vendas > 10000 and anos_empresa >= 1:
        bonus = 500.0
    else:
        bonus = 0.0
total = salario_base + comissao + bonus

print('\n--- Demonstrativo de Pagamento ---')
print(f'Salário base: R$ {salario_base:>10,.2f}')
print(f'Comissão (5%): R$ {comissao:>10,.2f}')
print(f'Bônus: R$ {bonus:>10,.2f}')
print('-' * 34)
print(f'TOTAL: R$ {total:>10,.2f}')
```

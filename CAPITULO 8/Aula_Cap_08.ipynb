```python
vendas = float(input('Total: R$'))
salario_inicial = float(input('Qual o valor do salario inicial?'))
anos_empresa = int(input('Quantos anos de empresa?'))
comissao = vendas * 0.05

if (vendas > 10000) and (anos_empresa >= 1):
  bonus = 500.00
else:
  bonus = 0.00

salario_final = salario_inicial + comissao + bonus

print(f'O salario inicial é R$ {salario_inicial:10.2f}')
print(f'A comissão é........R$ {comissao:10.2f}')
print(f'O bonus é>..........R$ {bonus:10.2f}')
print(f'O salario final é...R$ {salario_final:10.2f}')
```

#ENTRADAS:

A = ATENDIMENTO PERSONALIZADO

* A.1 = ANÁLISE DE DADOS DOS CLIENTES SOBRE COMPRAS ANTERIORES

* A.2 = ANÁLISE DO PERFIL DO CLIENTE E  DE SUAS DATAS COMEMORATIVAS

* A.3 = OUVIR MAIS E FALAR MENOS, CONVERSAR COM O CLIENTE PARA ENTENDER QUAIS SÃO SUAS DEMANDAS DO MOMENTO QUE PODEMOS ESTAR SUPRINDO

* A.4 = CASO NOSSO PRODUTO SEJA ADAPTÁVEL, VERIFICAR COMO PODEMOS ESTAR PERSONALIZANDO ELE DA MELHOR FORMA PARA O CLIENTE.

B = ATENDIMENTO HUMANIZADO

* B.1 = OUVIR MAIS E FALAR MENOS, CONVERSAR COM O CLIENTE PARA ENTENDER QUAL O "MOMENTO DELE" ATUALMENTE.

C = ATENDIMENTO QUE ATINJA OS 5 SENTIDOS HUMANOS, O CUMPRIMENTO DESSES 5 SENTIDOS EM UM ATENDIMENTO AUMENTARÁ SIGNIFICATIVAMENTE O PERCENTUAL DE UMA EXPERIENCIA AGRADÁVEL E MEMORÁVEL.

* C.1 = OLFATO : GARANTIR QUE O ESPAÇO TEM UM AROMA AGRADÁVEL

    * C.1.1 = DEFINIR UM AROMA PADRÃO, PARA SEMPRE QUE O CLIENTE SENTIR O CHEIRO ELE POSSA ESTÁ ATIVANDO SUA MEMÓRIA OLFATIVA E LEMBRANDO DA EMPRESA DE FORMA AGRADÁVEL.

* C.2 = VISÃO : TER UMA VISÃO AGRADÁVEL

    * C.2.1 = TER MUITO CUIDADO PARA MANTER: ORGANIZAÇÃO E LIMPEZA DO ESPAÇO.
    * C.2.2 = IMAGEM VISUAL DO AMBIENTE DEVE MANTER LIGAÇÃO DIRETA COM AS CORES DA LOGO DA EMPRESA.
    * C.2.3 = DEVE SE TER CUIDADO PARA O AMBIENTE ESTÁ EM HARMONIA COM A VIBE DA EMPRESA, E CUIDADO PARA NÃO CAUSAR UMA POLUIÇÃO VISUAL NO AMBIENTE.

* C.3 = TATO

    * C.3.1 = GARANTIR QUE OS ELEMENTOS DO AMBIENTE TENHAM PONTAS ARREDONDADAS, SEM ARESTAS QUE POSSAM INCOMODAR.
    * C.3.2 = GARANTIR A UNIFORMIDADE DE SUPERFICIES E ADEQUAR QUE CADA UMA SEJA APROPRIADA PARA CADA ESPAÇO DO AMBIENTE.

* C.4 = AUDIÇÃO

    * C.4.1 = GARANTIR QUE O ESPAÇO TENHA UM FUNDO MUSICAL AMBIENTE DE ACORDO COM A VIBE DA EMPRESA.
    * C.4.2 = GARANTIR QUE O FUNDO MUSICAL ESTEJA EM UMA ALTURA AGRADÁVEL PARA QUE AS PESSOAS CONSIGAM CONVERSAR SEM SEREM INCOMODADAS.

* C.5 = PALADAR

    * C.5.1 = FORNECER AO CLIENTE UMA EXPERIÊNCIA PARA AGRADAR O SEU PALADAR, COMO POR EXEMPLO UMA ENTRADINHA.

D = ATENDENTE

E = CLIENTE

F = AUX. DE SERVIÇOS GERAIS

G = AUX. DE COZINHA

I = CAIXA DE SOM

J = SALÃO DE ATENDIMENTO

K = DIFUSOR DE AROMAS


#PROCESSAMENTO:

PREPARAÇÃO DO AMBIENTE (SETUP SENSORIAL): O AUX. DE SERVIÇOS GERAIS (F) GARANTE A LIMPEZA E ORGANIZAÇÃO DO SALÃO DE ATENDIMENTO (J) SEGUINDO AS DIRETRIZES VISUAIS (C.2) E DE SEGURANÇA TÁTIL (C.3). O DIFUSOR DE AROMAS (K) É ACIONADO COM A FRAGRÂNCIA PADRÃO (C.1) E A CAIXA DE SOM (I) AJUSTADA NA PLAYLIST E VOLUME ADEQUADOS (C.4).

RECEPÇÃO E ACOLHIMENTO: O ATENDENTE (D) RECEBE O CLIENTE (E) APLICANDO A ESCUTA ATIVA DO ATENDIMENTO HUMANIZADO (B.1). EM PARALELO, O AUX. DE COZINHA (G) FORNECE A CORTESIA GASTRONÔMICA PARA ATIVAR O PALADAR (C.5).

DIAGNÓSTICO DE DEMANDA: O ATENDENTE (D) CRUZA AS INFORMAÇÕES OBTIDAS NA CONVERSA (A.3) COM OS DADOS PRÉVIOS DE COMPRAS (A.1) E DATAS COMEMORATIVAS (A.2) DO CLIENTE (E).

CUSTOMIZAÇÃO E ENTREGA: COM BASE NO DIAGNÓSTICO, O ATENDENTE (D) ADAPTA E PERSONALIZA O PRODUTO/SERVIÇO DA MELHOR FORMA (A.4) E REALIZA A ENTREGA FINAL, GARANTINDO QUE O CLIENTE SE SINTA ÚNICO.

#SAÍDA:

PRODUTO OU SERVIÇO ENTREGUE DE FORMA 100% PERSONALIZADA E ADAPTADA AO MOMENTO ATUAL DO CLIENTE.

UMA EXPERIÊNCIA DE CONSUMO MEMORÁVEL CONCLUÍDA, COM OS 5 SENTIDOS DO CLIENTE POSITIVAMENTE ESTIMULADOS.

HISTÓRICO E PERFIL DO CLIENTE ATUALIZADOS NO BANCO DE DADOS DA EMPRESA PARA FACILITAR INTERAÇÕES FUTURAS.

#CRITÉRIO DE SUCESSO:

MÉTRICA DE SATISFAÇÃO DIRETA: OBTER NOTA MÁXIMA (EX: 5/5 OU 10/10) NA PESQUISA DE CSAT (CUSTOMER SATISFACTION SCORE) APLICADA LOGO APÓS A EXPERIÊNCIA.

MÉTRICA DE RECOMENDAÇÃO: ATINGIR UM NPS (NET PROMOTER SCORE) NA ZONA DE EXCELÊNCIA (ACIMA DE 75%), COMPROVANDO QUE A EXPERIÊNCIA FOI BOA O SUFICIENTE PARA O CLIENTE INDICAR A EMPRESA.

TAXA DE RETENÇÃO: O CLIENTE REALIZAR UMA NOVA COMPRA, AGENDAMENTO OU INTERAÇÃO COM A MARCA DENTRO DOS PRÓXIMOS 90 DIAS.

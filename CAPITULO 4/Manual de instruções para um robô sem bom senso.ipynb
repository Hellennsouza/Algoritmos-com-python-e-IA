Este projeto é aberto: você escolhe uma tarefa simples do cotidiano e escreve um manual
de instruções tão detalhado que até um robô sem nenhum bom senso conseguiria executálo.

* Primeiro passo: Separe a louça suja que está na mesa da sala de jantar

* Segundo passo: Jogue o excesso de comida da louça no lixo que está do lado da pia da cozinha
* Terceiro passo: Coloque a louça suja que estava na sala de jantar na pia
* Quarto passo: Pegue um bombril que está na porta do armario inferior direito e coloque sobre a buxa de lavar prato que está na pia da cozinha.
* Quinto passo: Molhe a buxa com o brombril
* Sexto passo: Coloque o detergente que está na pia sobre a buxa de lavar prato
* Sétimo passo: Passe a buxa com o bombril em toda area da louça no sentido horário de modo a limpar a louça
* Oitavo passo: Enxague a louça com agua usando a torneira da pia
* Nono passo: Verifique que a louça está realmente limpa, caso contrário refaça todos os passos
Parabéns! A louça está limpa!

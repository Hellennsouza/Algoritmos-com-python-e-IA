#**1º Enunciado:**
Mostre, com um exemplo cada, as três formas de trazer recursos de um módulo para o programa: importando o módulo inteiro, importando uma função específica e importando com apelido.

````python
#1º Resposta:
# 1) Módulo inteiro: uso sempre com o prefixo math.
import math
print(math.sqrt(49))

# 2) Função específica: sqrt passa a existir "sozinha"
from math import sqrt
print(sqrt(81))

# 3) Com apelido: statistics passa a se chamar stats
import statistics as stats
valores = [10, 20, 30]
print(stats.mean(valores))
````

##1º Resposta (quando usar cada forma):
| Forma | Exemplo | Vantagem | Cuidado |
|---|---|---|---|
| Módulo inteiro | `import math` → `math.sqrt(49)` | Deixa claro de onde vem cada função | Escreve o prefixo toda vez |
| Função específica | `from math import sqrt` → `sqrt(81)` | Código mais curto | Quem lê não vê de onde a função veio |
| Apelido | `import statistics as stats` → `stats.mean(...)` | Nome curto e origem visível | Usar apelidos conhecidos (`pd`, `np`, `plt`) |

`stats.mean([10, 20, 30])` mostra `20` (e não `20.0`) porque a média é exata e o `statistics` mantém o tipo inteiro.

#**5º Enunciado:**
Simule 1000 lançamentos de um dado de seis faces, calcule média, mediana e desvio padrão, verifique se a média está próxima de 3.5 e exiba a frequência de cada face em um histograma textual.

````python
#5º Resposta:
import math
import random
import statistics as stats

random.seed(10)

resultados = []
for _ in range(1000):
    resultados.append(random.randint(1, 6))

media = stats.mean(resultados)
print(f'Média: {media:.2f}')
print(f'Mediana: {stats.median(resultados):.2f}')
print(f'Desvio padrão: {stats.pstdev(resultados):.2f}')
print(f'Média próxima de 3.5? {math.isclose(media, 3.5, abs_tol=0.2)}')

print('\nFrequência por face (cada █ = 10 lançamentos)')
total = 0
for face in range(1, 7):
    frequencia = resultados.count(face)
    total += frequencia
    print(f'{face}: {"█" * (frequencia // 10)} ({frequencia})')

assert total == 1000      # conferência: nenhuma face ficou de fora
print(f'Soma das frequências: {total}')
````

#**7º Enunciado:**
Explique por que o código abaixo imprime False, mesmo que 0.1 + 0.2 "pareça" igual a 0.3.

```python
x = 0.1 + 0.2
print(x == 0.3)
```

##7º Resposta:
O computador guarda números em **binário**. Assim como 1/3 vira 0,3333... em decimal e nunca termina, 0.1 e 0.2 não têm representação binária exata. A soma fica em `0.30000000000000004`, e o `==` compara igualdade **exata**, por isso dá `False`. Não é bug do Python. A regra é nunca comparar floats com `==`: use `math.isclose()`, que aceita uma diferença minúscula.

````python
#7º Resposta (demonstração):
import math

x = 0.1 + 0.2
print(x)
print(x == 0.3)
print(math.isclose(x, 0.3))
````

#**8º Enunciado:**
O código usa random.choices para sortear três pessoas de uma lista de três nomes. Identifique por que isso pode estar errado e como corrigir.

##8º Resposta:
`random.choices` sorteia **com reposição**: o nome sorteado "volta para a urna" e pode sair de novo. Para escolher três pessoas **diferentes**, o certo é `random.sample`, que sorteia **sem reposição**. A pergunta-chave é sempre: *pode repetir?*

````python
#8º Resposta (corrigido):
import random

random.seed(3)
nomes = ['Ana', 'Bruno', 'Carla']

com_reposicao = random.choices(nomes, k=3)     # pode repetir
sem_reposicao = random.sample(nomes, k=3)      # cada um só uma vez

print('Com reposição (choices):', com_reposicao)
print('Sem reposição (sample): ', sem_reposicao)
assert len(set(sem_reposicao)) == 3            # confirma que não houve repetição
````

#**9º Enunciado:**
A função abaixo calcula a média de uma lista usando stats.mean. Identifique o que acontece se a lista estiver vazia e proponha uma correção.

```python
def media_lista(valores):
    return stats.mean(valores)
```

##9º Resposta:
Com lista vazia, `stats.mean([])` gera `StatisticsError: mean requires at least one data point`, e o programa para com uma mensagem pouco clara para quem usa a função. A correção é verificar a lista **antes** de calcular e avisar com `raise ValueError('mensagem')`. Quem chama a função usa `try/except` para tratar o aviso sem travar.

````python
#9º Resposta (corrigido):
import statistics as stats

def media_lista(valores):
    """Calcula a média de uma lista não vazia."""
    if len(valores) == 0:
        raise ValueError('A lista não pode estar vazia.')
    return stats.mean(valores)

assert media_lista([10, 20, 30]) == 20
assert media_lista([5]) == 5

try:
    media_lista([])
except ValueError as erro:
    print('Erro tratado:', erro)
````

#**11º Enunciado:**
Reescreva o Exercício 10 usando pelo menos três funções (gerar_tempos, calcular_estatisticas e classificar_atendimento), cada uma com docstring e pelo menos dois testes com assert.

````python
#11º Resposta:
import math
import random
import statistics as stats


def gerar_tempos(qtd, minimo=2, maximo=20):
    """Gera uma lista com qtd tempos aleatórios de atendimento, em minutos."""
    if qtd <= 0:
        raise ValueError('A quantidade deve ser positiva.')
    if minimo > maximo:
        raise ValueError('O mínimo não pode ser maior que o máximo.')
    tempos = []
    for _ in range(qtd):
        tempos.append(random.uniform(minimo, maximo))
    return tempos


def calcular_estatisticas(tempos):
    """Retorna um dicionário com média, mediana, menor, maior e desvio padrão."""
    if len(tempos) == 0:
        raise ValueError('A lista de tempos não pode estar vazia.')
    return {
        'media': stats.mean(tempos),
        'mediana': stats.median(tempos),
        'menor': min(tempos),
        'maior': max(tempos),
        'desvio': stats.pstdev(tempos),
    }


def classificar_atendimento(media):
    """Classifica o atendimento pela média de tempo, em minutos."""
    if media <= 5:
        return 'excelente'
    if media <= 10:
        return 'adequado'
    return 'precisa de melhoria'


# ---------- Testes ----------
random.seed(1)
tempos_teste = gerar_tempos(50)
assert len(tempos_teste) == 50
for t in tempos_teste:
    assert 2 <= t <= 20

try:
    gerar_tempos(0)
    assert False, 'deveria ter dado ValueError'
except ValueError:
    pass

estat = calcular_estatisticas([2, 4, 6, 8])
assert math.isclose(estat['media'], 5.0)
assert math.isclose(estat['mediana'], 5.0)
assert estat['menor'] == 2 and estat['maior'] == 8

assert classificar_atendimento(5.0) == 'excelente'       # fronteira
assert classificar_atendimento(10.0) == 'adequado'       # fronteira
assert classificar_atendimento(10.5) == 'precisa de melhoria'
print('Todos os testes passaram.')

# ---------- Programa principal ----------
random.seed(2)
tempos = gerar_tempos(50)
estatisticas = calcular_estatisticas(tempos)
print(f"Média: {estatisticas['media']:.2f} | Mediana: {estatisticas['mediana']:.2f} | Desvio: {estatisticas['desvio']:.2f}")
print(f"Classificação: {classificar_atendimento(estatisticas['media'])}")
````

#**12º Enunciado:**
Leia a data de nascimento de uma pessoa e exiba: a idade aproximada em anos, o total de dias vividos, o dia da semana em que nasceu e quantos dias faltam para o próximo aniversário.

````python
#12º Resposta:
from datetime import date, datetime

DIAS_SEMANA = ['segunda-feira', 'terça-feira', 'quarta-feira', 'quinta-feira',
               'sexta-feira', 'sábado', 'domingo']


def ler_data(texto):
    """Converte 'DD/MM/AAAA' em um objeto date."""
    return datetime.strptime(texto.strip(), '%d/%m/%Y').date()


def calcular_idade(nascimento, hoje):
    """Idade em anos completos (desconta 1 se o aniversário deste ano ainda não chegou)."""
    idade = hoje.year - nascimento.year
    if (hoje.month, hoje.day) < (nascimento.month, nascimento.day):
        idade -= 1
    return idade


def total_dias_vividos(nascimento, hoje):
    """Dias entre o nascimento e hoje."""
    return (hoje - nascimento).days


def dia_da_semana(data):
    """Nome do dia da semana em português (weekday(): 0 = segunda)."""
    return DIAS_SEMANA[data.weekday()]


def dias_para_proximo_aniversario(nascimento, hoje):
    """Dias até o próximo aniversário (29/02 vira 28/02 em ano não bissexto)."""
    dia = nascimento.day
    if nascimento.month == 2 and dia == 29:
        dia = 28
    aniversario = date(hoje.year, nascimento.month, dia)
    if aniversario < hoje:
        aniversario = date(hoje.year + 1, nascimento.month, dia)
    return (aniversario - hoje).days


# ---------- Testes com datas fixas ----------
hoje_fixo = date(2026, 6, 13)
nasc = date(2005, 3, 15)
assert calcular_idade(nasc, hoje_fixo) == 21
assert total_dias_vividos(nasc, hoje_fixo) == 7760
assert dia_da_semana(nasc) == 'terça-feira'
assert dias_para_proximo_aniversario(nasc, hoje_fixo) == 275
assert dias_para_proximo_aniversario(date(2000, 6, 13), hoje_fixo) == 0     # aniversário hoje
assert calcular_idade(date(2000, 6, 14), hoje_fixo) == 25                    # faz 26 amanhã
print('Testes passaram!\n')

# ---------- Programa principal ----------
nascimento = ler_data(input('Data de nascimento (DD/MM/AAAA): '))
hoje = date.today()
print(f'Idade: {calcular_idade(nascimento, hoje)} anos')
print(f'Total de dias vividos: {total_dias_vividos(nascimento, hoje)}')
print(f'Dia da semana do nascimento: {dia_da_semana(nascimento)}')
print(f'Dias para o próximo aniversário: {dias_para_proximo_aniversario(nascimento, hoje)}')
````

##12º Resposta (observações):
- `strftime('%A')` devolve o dia em **inglês** no Colab ("Tuesday"). Para mostrar em português, usei `weekday()` (0 = segunda … 6 = domingo) com uma lista de nomes.
- A idade é calculada pelos anos completos, e não por `dias // 365`. Esse atalho erra perto do aniversário por causa dos anos bissextos.
- Quem nasceu em 29/02 comemora em 28/02 nos anos não bissextos; sem esse ajuste, `date(2026, 2, 29)` daria erro.

##14º Resposta — Mini-tutorial: módulo `collections` (classe `Counter`)

**Para que serve:** o `collections` traz estruturas de dados prontas além de lista e dicionário. A mais útil para análise de dados é o **`Counter`**, um "contador automático": você entrega uma lista e ele devolve quantas vezes cada item aparece, sem precisar montar o dicionário de contagem à mão.

**Funções úteis:**
| Recurso | O que faz |
|---|---|
| `Counter(lista)` | conta cada elemento da lista |
| `contador['x']` | quantas vezes `'x'` apareceu (0 se nunca apareceu, sem `KeyError`) |
| `contador.most_common(n)` | os `n` itens mais frequentes, do maior para o menor |
| `sum(contador.values())` | total de itens contados |
| `contador.update(nova_lista)` | soma novas ocorrências à contagem |

**Fonte:** documentação oficial — docs.python.org/pt-br/3/library/collections.html

````python
#14º Resposta — exemplo de código:
from collections import Counter

vendas_do_dia = ['café', 'pão de queijo', 'café', 'suco', 'café', 'bolo',
                 'pão de queijo', 'suco', 'café', 'pão de queijo']

contagem = Counter(vendas_do_dia)

print('Contagem completa:', dict(contagem))
print('Cafés vendidos:', contagem['café'])
print('Tapiocas vendidas:', contagem['tapioca'])      # não existe -> 0, sem erro

print('\nTop 3 produtos:')
for posicao, (produto, qtd) in enumerate(contagem.most_common(3), 1):
    print(f'{posicao}º {produto:<15} {"█" * qtd} ({qtd})')

# chegaram as vendas da tarde
contagem.update(['bolo', 'bolo', 'suco'])
print('\nDepois da tarde:', contagem.most_common(2))
print('Total de itens vendidos:', sum(contagem.values()))

# comparação: o mesmo resultado "na mão", como no Cap. 14
manual = {}
for item in vendas_do_dia:
    manual[item] = manual.get(item, 0) + 1
assert manual == dict(Counter(vendas_do_dia))
````

#**15º Enunciado:**
Solicite a uma IA uma função de histograma textual e audite a resposta verificando cinco pontos: tratamento da última faixa, comportamento quando todos os valores são iguais, validação de n_bins, clareza dos nomes e presença de testes.

````python
#15º Resposta — primeira versão gerada pela IA (Claude), antes da auditoria:
def hist(a, n=5):
    mn, mx = min(a), max(a)
    w = (mx - mn) / n
    for i in range(n):
        lo, hi = mn + i * w, mn + (i + 1) * w
        c = len([x for x in a if lo <= x < hi])
        print(f'[{lo:.2f}, {hi:.2f}): ' + '#' * c)

print('Teste 1 — valores 1 a 5 em 2 faixas (esperado: 5 valores no total)')
hist([1, 2, 3, 4, 5], 2)

for teste, args in [('valores iguais', ([3, 3, 3], 3)), ('n = 0', ([1, 2, 3], 0)), ('lista vazia', ([], 2))]:
    try:
        hist(*args)
    except Exception as erro:
        print(f'Teste {teste} -> {type(erro).__name__}: {erro}')
````

| Critério | O que encontrei na versão da IA | Correção |
|---|---|---|
| Última faixa | Usa `<` em todas: o **5 (maior valor) ficou de fora** (contou 4 de 5) | Última faixa com `<=` |
| Valores iguais | `[3, 3, 3]` imprime 3 faixas vazias em vez de avisar; a largura vira 0 e o 3 não entra em nenhuma | Detectar com `math.isclose(minv, maxv)` |
| n_bins inválido | `n = 0` gera `ZeroDivisionError` sem mensagem clara | `raise ValueError` se `n_bins <= 0` |
| Nomes | `a`, `n`, `w`, `lo`, `hi`, `c` não dizem nada | `amostras`, `n_bins`, `largura`, `inicio`, `fim`, `conta` |
| Testes | Nenhum teste; lista vazia gera `ValueError` do `min()` sem explicação | Função devolve as contagens e é testada com `assert` |

````python
#15º Resposta — versão corrigida após a auditoria:
import math


def contar_faixas(amostras, n_bins=5):
    """Devolve uma lista de (inicio, fim, conta) para cada faixa do histograma."""
    if len(amostras) == 0:
        raise ValueError('A lista não pode estar vazia.')
    if n_bins <= 0:
        raise ValueError('n_bins deve ser maior que zero.')

    minv = min(amostras)
    maxv = max(amostras)
    if math.isclose(minv, maxv):                  # todos iguais: uma faixa só
        return [(minv, maxv, len(amostras))]

    largura = (maxv - minv) / n_bins
    faixas = []
    for b in range(n_bins):
        inicio = minv + b * largura
        fim = inicio + largura
        conta = 0
        for x in amostras:
            if b == n_bins - 1:                   # última faixa inclui o limite superior
                if inicio <= x <= maxv:
                    conta += 1
            elif inicio <= x < fim:
                conta += 1
        faixas.append((inicio, fim, conta))
    return faixas


def histograma_textual(amostras, n_bins=5):
    """Imprime o histograma (a contagem fica em contar_faixas)."""
    faixas = contar_faixas(amostras, n_bins)
    if len(faixas) == 1 and math.isclose(faixas[0][0], faixas[0][1]):
        print(f'Todos os {faixas[0][2]} valores são iguais a {faixas[0][0]}.')
        return
    for i in range(len(faixas)):
        inicio, fim, conta = faixas[i]
        fecha = ']' if i == len(faixas) - 1 else ')'
        print(f'[{inicio:.2f} - {fim:.2f}{fecha}: {"█" * conta} ({conta})')


# ---------- Testes ----------
faixas = contar_faixas([1, 2, 3, 4, 5], 2)
assert [f[2] for f in faixas] == [2, 3]
total = 0
for f in faixas:
    total += f[2]
assert total == 5                                  # nenhum valor ficou de fora
assert contar_faixas([3, 3, 3], 3) == [(3, 3, 3)]
for invalido in [([1, 2, 3], 0), ([1, 2, 3], -2), ([], 2)]:
    try:
        contar_faixas(*invalido)
        assert False
    except ValueError as erro:
        print('Recusado corretamente:', erro)
print('Testes passaram!\n')

histograma_textual([1, 2, 3, 4, 5], 2)
print('---')
histograma_textual([3, 3, 3], 3)
print('---')
histograma_textual([2.5, 3.1, 4.8, 5.0, 5.2, 6.7, 7.0, 7.4, 8.8, 9.9], 4)
````

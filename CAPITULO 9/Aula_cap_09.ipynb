Se estiver chovendo, levo guarda-chuva. Se estiver frio e chovendo, levo o casaco também. Senão, vou de camiseta.

```python
chuva = input("Está chovendo? ")
temperatura = input("Está frio? ")
```

```python
if chuva == "sim":
    if temperatura == "sim":
        print("Levar casaco e guarda-chuva")
    else:
        print("Levar guarda-chuva")
else:
    if temperatura == "sim":
        print("Levar casaco")
    else:
        print("Vou de camiseta")
```

Faça um algoritmo para verificar se um número é positivo ou negativo... Caso seja positivo... Verifique se é par ou impar.

```python
numero = int(input("Digite um número inteiro: "))
if numero >= 0:
    if numero % 2 == 0:
        print("O número é positivo e par")
    else:
        print("O número é positivo e ímpar")
else:
    print("O número é negativo!")
```

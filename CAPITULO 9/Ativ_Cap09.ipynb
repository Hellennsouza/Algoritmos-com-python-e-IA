#**1º Enunciado:** - Maior Idade
Peça a idade da pessoa e exiba se ela é maior de idade (18 anos ou mais) ou menor de
idade.

```python
#1º Resposta
idade = int(input("Digite a idade: "))
if idade >= 18:
    print("Maior de idade")
else:
    print("Menor de idade")
```

#**3º Enunciado:**
Peça a temperatura e classifique-a em três faixas: abaixo de 15°C é frio, de 15°C a 25°C
(sem incluir 25°C) é agradável, e 25°C ou mais é quente.

```python
#3º Resposta:
temperatura = float(input("Digite a temperatura: "))
if temperatura < 15:
    print("Frio")
elif temperatura < 25:
    print("Agradável")
else:
    print("Quente")
```

#**4º Enunciado**
Peça o valor de uma compra. Se o valor for maior que R$ 200,00, aplique um desconto de
10%; caso contrário, o valor final é igual ao valor original. Exiba o valor original e o valor
final, com duas casas decimais.

```python
valor_compra = float(input("Digite o valor da compra: "))
if valor_compra > 200:
    valor_final = valor_compra * 0.9
else:
    valor_final = valor_compra
print(f"Valor original: R$ {valor_compra:.2f}")
print(f"Valor final: R$ {valor_final:.2f}")
```

#**5º Enunciado:**

Peça o total comprado por um cliente no mês e classifique-o em uma de três categorias:
cliente ouro (total >= R$ 1.000), cliente prata (total >= R$ 500) ou cliente bronze (qualquer
outro valor).

```python
total_compras = float(input("Digite o total comprado por um cliente no mês: "))
if total_compras >= 1000:
    categoria = "ouro"
elif total_compras >= 500:
    categoria = "prata"
else:
    categoria = "bronze"
print("Este cliente esta na categoria:",categoria)
```

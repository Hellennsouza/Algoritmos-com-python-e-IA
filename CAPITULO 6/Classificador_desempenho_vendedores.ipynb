O projeto pede um sistema de classificação de desempenho de vendedores, usando as três
faixas já conhecidas, representado nas cinco formas trabalhadas neste capítulo.



1. Descrição do problema
Ler o valor total vendido por um vendedor no mês. Se o valor for menor que R$ 5.000,
classificar como "Abaixo da meta". Se estiver entre R$ 5.000 e R$ 10.000, incluindo os dois
limites, classificar como "Dentro da meta". Se for maior que R$ 10.000, classificar como
"Acima da meta".

flowchart TD

*    diga [Abaixo da meta] por (2) segundos

```python
valor_vendido = float(input('Valor vendido no mês: '))
if valor_vendido < 5000:
print('Abaixo da meta')
elif valor_vendido <= 10000:
print('Dentro da meta')
else:
print('Acima da meta')
```
